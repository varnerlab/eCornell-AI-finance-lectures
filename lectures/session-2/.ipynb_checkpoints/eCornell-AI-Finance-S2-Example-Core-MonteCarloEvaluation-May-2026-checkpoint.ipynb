# Example: Monte Carlo Evaluation of the Rebalancing Engine

The [rebalancing scorecard notebook](eCornell-AI-Finance-S2-Example-Core-RebalancingEngineScorecard-May-2026.ipynb) priced the engine on the actual calendar-2025 path. This example asks the forward-looking question: **what does the engine actually deliver versus passive baselines using future synthetic paths once we strip out a known SIM artifact, and where is its real value-add?**

> __Learning Objectives:__
>
> By the end of this example, you will be able to:
> * __Decompose a dynamic strategy's synthetic return:__ Decompose a dynamic strategy's simulated return into the share attributable to passive allocation, daily rebalancing, and sentiment tilt. Recognize the share that is a generator artifact (iid residuals inflate diversification return) and discount it before reporting forward edge.
> * __Test the engine's universe sensitivity:__ Re-run the same allocator on a structurally different asset basket and read how the universe shifts the strategy's risk-return profile. Distinguish edge that comes from the allocator's mechanics from edge that comes from the alignment between calibrated parameters and the deployment regime.
> * __Demystify daily rebalancing with a trade-level diagnostic:__ Translate per-day allocator outputs into a trade-level diagnostic that exposes dollar turnover, transaction cost, trigger fires, and concentration. Make the engine's mechanical activity legible so behavior can be debugged and explained, not assumed.

Let's dive in!

___

### Implementation
We define six helper functions used by Task 1 to build the per-path noise-injected engine and static Cobb-Douglas baselines, compute tail-risk metrics, and render the confidence-cone figure.

> __`draw_noisy_sim_params(sim_params_pt, bootstrap, rng)`__
>
> Draw a per-path realization of the SIM parameters by adding Gaussian noise at bootstrap-SE scale to each ticker's $\alpha$ and $\beta$ point estimates. Returns a `Dict{String,Tuple{Float64,Float64,Float64}}` with the same shape as `sim_params`: ticker $\mapsto (\alpha_i + \varepsilon_\alpha \sigma_{\alpha,i},\ \beta_i + \varepsilon_\beta \sigma_{\beta,i},\ \sigma_{\varepsilon,i})$. $\sigma_\varepsilon$ is held at the point estimate (bootstrap quantifies uncertainty in the regression coefficients, not in the residual scale).

> __`backtest_engine_noisy(scenario, tickers, sim_params_pt, bootstrap, rules_params, rng; B₀, offset, cost_bps)`__
>
> Path-by-path rebalancing-engine backtest with **fresh noisy SIM parameters per path**. For each path: recover market growth and EMA-crossover $\lambda$, draw `np = draw_noisy_sim_params(...)`, build a [`MyRebalancingContextModel`](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session2/#eCornellAIFinance.MyRebalancingContextModel) with those noisy params, run [the `run_rebalancing_engine(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session2/#eCornellAIFinance.run_rebalancing_engine) with `cost_bps` applied at each rebalance, and collect per-path final wealth, drawdown, and a CC-excess Sharpe. Returns a [`MyBacktestResult`](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.MyBacktestResult) with the same shape as [the `backtest_engine(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.backtest_engine) so downstream cells are unchanged.

> __`backtest_static_cd_noisy(scenario, tickers, sim_params_pt, bootstrap, lambda_mode, rng; ...)`__
>
> The `backtest_static_cd` pattern with the same per-path α/β noise injection. Fits $\gamma$ once at the buy-in day using the path's noisy SIM parameters and a static $\lambda$ (`:neutral` or `:warmup_avg`), holds the resulting shares fixed, and deducts the same buy-in cost the engine pays at its first rebalance so the two strategies are compared on equal footing.

> __`tail_metrics(r, B0, discount, Δt, rf_growth_pct)`__
>
> Produces one scorecard row as a `NamedTuple` with terminal-wealth quantiles, VaR and CVaR on the 5% tail (with an analytical standard error), drawdown quantiles, median Sharpe, and NPV columns. All wealth and NPV quantities are reported as W₀-scaled ratios (column suffix `_W0`) so the table reads as multiples of initial capital, matching the rest of the course. `Med_Sharpe` is recomputed per path from `r.wealth_paths` using the continuously-compounded excess-growth convention ($g_\text{CC} - g_f$) / annualized log-return vol, so every strategy row is in the same units regardless of which backtest produced the result.

> __`cones(W)`__
>
> Takes an `(n_steps+1) × n_paths` scaled-wealth matrix and returns a `NamedTuple` of pointwise empirical quantile bands (`p005, p025, p160, p500, mean, p840, p975, p995`) for the 68%, 95%, and 99% confidence cones used in the Task 1 tail-shape figure. Empirical rather than Gaussian $\pm k\sigma$ so the generative model futures' heavy tails are rendered honestly.

> __`panel(W, c, worst_idx, title_str, days, n_tail, rf_multiple; show_ylabel)`__
>
> Builds one panel of the Task 1 two-panel tail-shape figure: three nested empirical confidence cones behind the median and mean trajectories, the 5% worst paths overlaid in red, and reference hlines at initial wealth (1.0) and the risk-free terminal multiple. Used to compare the S1 min-var buy-and-hold against the Cobb-Douglas engine on a shared y-axis.

The six helper definitions appear in the code block below.

```julia
"""
    backtest_engine_frozen(scenario, tickers, sim_params_pt, rules_params;
        B₀, offset, L_short, L_long, L_growth, GAIN, epsilon, cost_bps) -> MyBacktestResult

Path-by-path rebalancing-engine backtest with the allocator reading the frozen
S1 point-estimate `sim_params_pt` on every path. `cost_bps` is applied inside
`run_rebalancing_engine` on every rebalance day.
"""
function backtest_engine_frozen(scenario::MyBacktestScenario,
    tickers::Vector{String},
    sim_params_pt::Dict{String,Tuple{Float64,Float64,Float64}},
    rules_params::NamedTuple;
    B₀::Float64 = 10_000.0, offset::Int = 84,
    L_short::Int = 21, L_long::Int = 63, L_growth::Int = 10,
    GAIN::Float64 = 10.0, epsilon::Float64 = 0.1,
    cost_bps::Float64 = 0.0)::MyBacktestResult

    Δt = 1.0 / 252.0;
    n_paths   = scenario.n_paths;
    n_steps   = scenario.n_steps;
    N         = length(tickers);
    n_trading = n_steps - offset;

    final_wealth  = zeros(n_paths);
    max_drawdowns = zeros(n_paths);
    sharpe_ratios = zeros(n_paths);
    wealth_paths  = zeros(n_trading + 1, n_paths);

    for p ∈ 1:n_paths
        mkt    = scenario.market_paths[p, :];
        gm_raw = compute_market_growth(mkt; Δt = Δt);
        gm_e   = compute_ema(gm_raw; window = L_growth);
        ema_s  = compute_ema(mkt; window = L_short);
        ema_l  = compute_ema(mkt; window = L_long);
        λ      = compute_lambda(ema_s, ema_l; G = GAIN);

        pmat = zeros(n_steps, N + 1);
        for t ∈ 1:n_steps
            pmat[t, 1] = t;
        end
        for k ∈ 1:N
            pmat[:, k + 1] = scenario.price_paths[p, :, k];
        end

        ctx = build(MyRebalancingContextModel, (
            B = B₀, tickers = tickers, marketdata = pmat,
            marketfactor = gm_e, sim_parameters = sim_params_pt,
            lambda = 0.0, Δt = Δt, epsilon = epsilon,
        ));
        rules = build(MyTriggerRules, (
            max_drawdown = rules_params.max_drawdown,
            max_turnover = rules_params.max_turnover,
            rebalance_schedule = ones(Int, n_trading),
        ));
        results = run_rebalancing_engine(ctx, rules, λ;
            offset = offset, allocator = :cobb_douglas, cost_bps = cost_bps);
        wealth = compute_wealth_series(results, pmat, tickers; offset = offset);

        wealth_paths[:, p] .= wealth;
        final_wealth[p]  = wealth[end];
        # Running high-water mark: peak[t] = max(W_0, ..., W_t).
        peak             = accumulate(max, wealth);
        # Drawdown fraction: (peak − current) / peak, in [0, 1); `max_drawdowns[p]` is the worst over the path.
        max_drawdowns[p] = maximum((peak .- wealth) ./ peak);
        # Annualized volatility (finance convention): std of daily simple returns × √252.
        vol              = std(diff(wealth) ./ wealth[1:end-1]) * sqrt(252);
        mean_ret         = (wealth[end] / wealth[1] - 1.0);
        sharpe_ratios[p] = vol > 1e-6 ? mean_ret / vol : 0.0;
    end

    result = MyBacktestResult();
    result.scenario_label = scenario.label;
    result.strategy_label = "Cobb-Douglas Engine (cost=$(cost_bps) bps)";
    result.final_wealth   = final_wealth;
    result.max_drawdowns  = max_drawdowns;
    result.sharpe_ratios  = sharpe_ratios;
    result.wealth_paths   = wealth_paths;
    return result;
end;

"""
    backtest_static_cd_frozen(scenario, tickers, sim_params_pt, lambda_mode;
        B₀, offset, L_short, L_long, L_growth, GAIN, epsilon, cost_bps) -> MyBacktestResult

Static Cobb-Douglas buy-and-hold using the frozen S1 point estimates.
Allocates once at the buy-in day under the chosen `lambda_mode` (`:neutral`
or `:warmup_avg`), deducts the same per-trade cost the engine pays on its
first rebalance, then holds the shares through the active window.
"""
function backtest_static_cd_frozen(scenario::MyBacktestScenario,
    tickers::Vector{String},
    sim_params_pt::Dict{String,Tuple{Float64,Float64,Float64}},
    lambda_mode::Symbol;
    B₀::Float64 = 10_000.0, offset::Int = 84,
    L_short::Int = 21, L_long::Int = 63, L_growth::Int = 10,
    GAIN::Float64 = 10.0, epsilon::Float64 = 0.1,
    cost_bps::Float64 = 0.0)::MyBacktestResult

    Δt = 1.0 / 252.0;
    n_paths   = scenario.n_paths;
    n_steps   = scenario.n_steps;
    N         = length(tickers);
    n_trading = n_steps - offset;

    final_wealth  = zeros(n_paths);
    max_drawdowns = zeros(n_paths);
    sharpe_ratios = zeros(n_paths);
    wealth_paths  = zeros(n_trading + 1, n_paths);

    B_eff = B₀ * (1.0 - cost_bps / 10_000.0);

    for p ∈ 1:n_paths
        mkt    = scenario.market_paths[p, :];
        gm_raw = compute_market_growth(mkt; Δt = Δt);
        gm_e   = compute_ema(gm_raw; window = L_growth);

        λ_static = if lambda_mode == :neutral
            0.0
        elseif lambda_mode == :warmup_avg
            ema_s = compute_ema(mkt; window = L_short);
            ema_l = compute_ema(mkt; window = L_long);
            λ_raw = compute_lambda(ema_s, ema_l; G = GAIN);
            mean(λ_raw[(L_long + 1):offset])
        else
            error("Unknown lambda_mode: $(lambda_mode). Use :neutral or :warmup_avg.")
        end

        buyin_day = offset + 1;
        gm_t = gm_e[min(buyin_day, length(gm_e))];
        γ = compute_preference_weights(sim_params_pt, tickers, gm_t, λ_static);

        prices_buyin = [scenario.price_paths[p, buyin_day, k] for k ∈ 1:N];
        problem = build(MyCobbDouglasChoiceProblem, (
            gamma = γ, prices = prices_buyin, B = B_eff, epsilon = epsilon,
        ));
        (shares, cash) = allocate_cobb_douglas(problem);

        wealth = zeros(n_trading + 1);
        for d ∈ 0:n_trading
            day = min(buyin_day + d, n_steps);
            wealth[d + 1] = cash + sum(shares[k] * scenario.price_paths[p, day, k] for k ∈ 1:N);
        end
        wealth_paths[:, p] .= wealth;

        final_wealth[p]  = wealth[end];
        peak             = accumulate(max, wealth);
        max_drawdowns[p] = maximum((peak .- wealth) ./ peak);
        vol              = std(diff(wealth) ./ wealth[1:end-1]) * sqrt(252);
        mean_ret         = (wealth[end] / wealth[1] - 1.0);
        sharpe_ratios[p] = vol > 1e-6 ? mean_ret / vol : 0.0;
    end

    result = MyBacktestResult();
    result.scenario_label = scenario.label;
    result.strategy_label = "Static CD ($(lambda_mode) λ, cost=$(cost_bps) bps)";
    result.final_wealth   = final_wealth;
    result.max_drawdowns  = max_drawdowns;
    result.sharpe_ratios  = sharpe_ratios;
    result.wealth_paths   = wealth_paths;
    return result;
end;

"""
    apply_bias_correction(r::MyBacktestResult, bias_pct_per_yr, Δt) -> MyBacktestResult

Apply a uniform per-step exponential drag to every wealth path so that each
path's continuously-compounded growth rate is reduced by `bias_pct_per_yr` /
year. The correction is

    W_corr[p, t] = W[p, t] · exp(-bias · (t-1) · Δt),

where `bias = bias_pct_per_yr / 100`. Per-path rank order is preserved
(rank correlation = 1 with the input); the entire distribution is location-
shifted in CCGR space. Used to anchor the synthetic engine's central tendency
to a realized reference (here the 2025 path) so the corrected percentile
structure characterizes uncertainty around a credible center rather than a
SIM-inflated point. See the decomposition cell below for justification.
"""
function apply_bias_correction(r::MyBacktestResult, bias_pct_per_yr::Float64,
        Δt::Float64)::MyBacktestResult
    n_t, n_p = size(r.wealth_paths);
    drag = exp.(-bias_pct_per_yr / 100.0 .* (0:(n_t-1)) .* Δt);
    Wc = r.wealth_paths .* drag;

    final_wealth_c  = Wc[end, :];
    max_drawdowns_c = zeros(n_p);
    sharpe_ratios_c = zeros(n_p);
    for p ∈ 1:n_p
        wealth = Wc[:, p];
        peak = accumulate(max, wealth);
        max_drawdowns_c[p] = maximum((peak .- wealth) ./ peak);
        vol = std(diff(wealth) ./ wealth[1:end-1]) * sqrt(252);
        mean_ret = (wealth[end] / wealth[1] - 1.0);
        sharpe_ratios_c[p] = vol > 1e-6 ? mean_ret / vol : 0.0;
    end

    rc = MyBacktestResult();
    rc.scenario_label = r.scenario_label;
    rc.strategy_label = r.strategy_label * " [bias-corrected $(round(bias_pct_per_yr, digits=2))pp/yr]";
    rc.final_wealth   = final_wealth_c;
    rc.max_drawdowns  = max_drawdowns_c;
    rc.sharpe_ratios  = sharpe_ratios_c;
    rc.wealth_paths   = Wc;
    return rc;
end;

"""
    tail_metrics(r::MyBacktestResult, B0::Float64, discount::Float64,
                 Δt::Float64, rf_growth_pct::Float64) -> NamedTuple

Compute one scorecard row from a per-path `MyBacktestResult`.

`discount` is the NPV discount factor `exp(-g_f * T_active)` and must be the same
value for every strategy so NPV columns are directly comparable. `rf_growth_pct`
is the risk-free continuously-compounded growth rate in %/yr; it's subtracted from
each path's CC growth when computing the per-path Sharpe, so the returned
`Med_Sharpe` is in excess-growth units regardless of the convention baked into
the upstream `MyBacktestResult.sharpe_ratios` field.
"""
function tail_metrics(r::MyBacktestResult, B0::Float64, discount::Float64,
                      Δt::Float64, rf_growth_pct::Float64)

    wealth = r.final_wealth;
    # NPV per path = discounted terminal wealth minus initial: W_T · e^(-g_f·T) − B₀.
    npv    = wealth .* discount .- B0;
    n      = length(wealth);
    n_tail = max(1, floor(Int, 0.05 * n));

    sorted_wealth = sort(wealth);
    wealth_tail   = sorted_wealth[1:n_tail];
    cvar5         = mean(wealth_tail);
    cvar5_se      = (length(wealth_tail) > 1) ? std(wealth_tail) / sqrt(n_tail) : 0.0;

    sorted_npv = sort(npv);
    npv_tail   = sorted_npv[1:n_tail];
    cvar5_npv  = mean(npv_tail);

    n_paths = length(wealth);
    sharpes = Vector{Float64}(undef, n_paths);
    for p ∈ 1:n_paths
        w = r.wealth_paths[:, p];
        horizon_yr = (length(w) - 1) * Δt;
        growth_pct = log(w[end] / w[1]) / horizon_yr * 100;
        log_rets   = diff(log.(w));
        vol_pct    = std(log_rets) * sqrt(252) * 100;
        excess     = growth_pct - rf_growth_pct;
        sharpes[p] = vol_pct > 1e-6 ? excess / vol_pct : 0.0;
    end

    return (
        Portfolio       = r.strategy_label,
        Med_W_W0        = round(median(wealth) / B0,                  digits=4),
        VaR5_W_W0       = round(quantile(wealth, 0.05) / B0,          digits=4),
        CVaR5_W_W0      = round(cvar5 / B0,                           digits=4),
        CVaR5_SE_W_W0   = round(cvar5_se / B0,                        digits=4),
        Med_DD_pct      = round(median(r.max_drawdowns) * 100,        digits=2),
        P95_DD_pct      = round(quantile(r.max_drawdowns, 0.95) * 100, digits=2),
        Med_Sharpe      = round(median(sharpes),                      digits=3),
        Med_NPV_W0      = round(median(npv) / B0,                     digits=4),
        CVaR5_NPV_W0    = round(cvar5_npv / B0,                       digits=4),
        P_NPV_neg_pct   = round(mean(npv .< 0) * 100,                 digits=1),
    );
end;

"""
    cones(W::Matrix{Float64}) -> NamedTuple

Compute pointwise empirical quantile bands of a scaled-wealth matrix
(shape: `(T, n_paths)`) for the 68%, 95%, and 99% confidence cones used
by the Task 1 tail-shape figure.
"""
function cones(W::Matrix{Float64})
    T = size(W, 1);
    return (
        p005 = [quantile(W[t, :], 0.005) for t ∈ 1:T],
        p025 = [quantile(W[t, :], 0.025) for t ∈ 1:T],
        p160 = [quantile(W[t, :], 0.160) for t ∈ 1:T],
        p500 = [median(W[t, :])           for t ∈ 1:T],
        mean = [mean(W[t, :])             for t ∈ 1:T],
        p840 = [quantile(W[t, :], 0.840) for t ∈ 1:T],
        p975 = [quantile(W[t, :], 0.975) for t ∈ 1:T],
        p995 = [quantile(W[t, :], 0.995) for t ∈ 1:T],
    );
end;

"""
    panel(W, c, worst_idx, title_str, days, n_tail, rf_multiple; show_ylabel=true) -> Plots.Plot

Build one tail-shape panel: three nested empirical confidence cones (68/95/99%)
behind the median and mean trajectories, with the 5% worst paths overlaid in
red and reference hlines at initial wealth (1.0) and the risk-free terminal
multiple.
"""
function panel(W::Matrix{Float64}, c, worst_idx, title_str::String,
               days, n_tail::Int, rf_multiple::Float64;
               show_ylabel::Bool = true)
    pl = plot(days, c.p995;
        fillrange = c.p005, fillalpha = 0.25, fillcolor = :lightskyblue,
        lw = 0.8, lc = :gray50, label = "99% band (P0.5:P99.5)",
        xlabel = "Trading Day (after 84-day warmup)",
        ylabel = show_ylabel ? "Scaled Wealth (W_t / W₀)" : "",
        fontsize = 16,
        left_margin = 12Plots.mm, right_margin = 6Plots.mm,
        top_margin = 4Plots.mm, bottom_margin = 8Plots.mm,
        size = (900, 520));
    plot!(pl, days, c.p005; lw = 0.8, lc = :gray50, label = "");
    plot!(pl, days, c.p975;
        fillrange = c.p025, fillalpha = 0.45, fillcolor = :lightskyblue,
        lw = 0.8, lc = :gray50, label = "95% band (P2.5:P97.5)");
    plot!(pl, days, c.p025; lw = 0.8, lc = :gray50, label = "");
    plot!(pl, days, c.p840;
        fillrange = c.p160, fillalpha = 0.75, fillcolor = :lightskyblue,
        lw = 0.8, lc = :gray50, label = "68% band (P16:P84)");
    plot!(pl, days, c.p160; lw = 0.8, lc = :gray50, label = "");
    for (j, idx) ∈ enumerate(worst_idx)
        plot!(pl, days, W[:, idx];
            lw = 1, c = :red, alpha = 0.20,
            label = (j == 1 ? "Worst $(n_tail) paths (5% tail)" : ""));
    end
    plot!(pl, days, c.p500; lw = 3, c = :gray20, label = "Median (all paths)");
    plot!(pl, days, c.mean; lw = 2, ls = :dot, c = :black, label = "Mean (all paths)");
    hline!(pl, [1.0];         lw = 2, ls = :dash, c = :black, alpha = 0.6,
        label = "Initial wealth (1.0)");
    hline!(pl, [rf_multiple]; lw = 2, ls = :dash, c = :seagreen,
        label = "Risk-Free terminal ($(round(rf_multiple, digits=3)))");
    plot!(pl, bg = "gray95",
        background_color_outside = "white",
        framestyle = :box,
        fg_legend = :transparent,
        legend = :topleft);
    title!(pl, title_str)
    return pl;
end;

"""
    trade_diagnostics(results::Dict{Int,MyRebalancingResult},
        marketdata::Matrix{Float64}, tickers::Vector{String};
        offset::Int = 84) -> NamedTuple

Extract per-day mechanical diagnostics from a `run_rebalancing_engine` result
dictionary: dollar value traded each day, total transaction cost paid (at the
caller-known cost_bps), boolean per-day trigger-fire vector (defined as `shares
all-zero today and at least one non-zero yesterday`), and per-day top-3 ticker
concentration as a fraction of total portfolio value. Used by Task 3.
"""
function trade_diagnostics(results::Dict{Int,MyRebalancingResult},
    marketdata::Matrix{Float64}, tickers::Vector{String};
    offset::Int = 84, cost_bps::Float64 = 5.0)

    K = length(tickers);
    T = maximum(keys(results));
    daily_traded   = zeros(T);
    daily_cost     = zeros(T);
    triggered      = falses(T);
    top3_concen    = zeros(T + 1);  # day 0..T
    top3_tickers   = Vector{Tuple{String,String,String}}(undef, T + 1);

    # --- Day 0: initial allocation ---
    r0 = results[0];
    initial_trade = sum(r0.shares[i] * marketdata[offset, i + 1] for i ∈ 1:K);
    # store as "day 1's incoming trade" via daily_traded[1]? Better: keep separate.

    # --- Per-day loop ---
    for t ∈ 1:T
        actual_day = offset + t;
        prev = results[t - 1];
        cur  = results[t];

        # dollar value traded today: sum |Δshares × price|
        traded_today = 0.0;
        for i ∈ 1:K
            traded_today += abs(cur.shares[i] - prev.shares[i]) * marketdata[actual_day, i + 1];
        end
        daily_traded[t] = traded_today;
        daily_cost[t]   = (cost_bps / 10_000.0) * traded_today;

        # trigger fired: shares all near-zero today and not all near-zero yesterday
        triggered[t] = all(abs.(cur.shares) .< 1e-9) && any(abs.(prev.shares) .> 1e-9);
    end

    # --- Per-day concentration + cash fraction (day 0..T) ---
    cash_frac = zeros(T + 1);
    total_wealth = zeros(T + 1);
    for t ∈ 0:T
        actual_day = offset + t;
        r = results[t];
        share_val = [r.shares[i] * marketdata[actual_day, i + 1] for i ∈ 1:K];
        total_val = sum(share_val) + r.cash;
        total_wealth[t + 1] = total_val;
        cash_frac[t + 1]    = total_val > 0 ? r.cash / total_val : 0.0;
        if total_val > 0
            sorted_idx = sortperm(share_val, rev = true);
            top3_concen[t + 1] = sum(share_val[sorted_idx[1:min(3, K)]]) / total_val;
            t1 = K >= 1 ? tickers[sorted_idx[1]] : "";
            t2 = K >= 2 ? tickers[sorted_idx[2]] : "";
            t3 = K >= 3 ? tickers[sorted_idx[3]] : "";
            top3_tickers[t + 1] = (t1, t2, t3);
        else
            top3_concen[t + 1] = 0.0;
            top3_tickers[t + 1] = ("", "", "");
        end
    end

    return (
        T               = T,
        daily_traded    = daily_traded,
        daily_cost      = daily_cost,
        triggered       = triggered,
        top3_concen     = top3_concen,
        top3_tickers    = top3_tickers,
        cash_frac       = cash_frac,
        total_wealth    = total_wealth,
        initial_trade   = initial_trade,
        initial_cost    = (cost_bps / 10_000.0) * initial_trade,
    );
end;
```

___
## Task 1: Build the Noise-Injected Tail-Risk Scorecard
In this task, we generate one multi-path $\times$ 336-day generative SIM scenario, run seven strategies across it (the engine, two static-CD variants, three buy-and-hold baselines, and a risk-free reference) on a shared 84-day warmup plus 252-day active window, and assemble a distributional scorecard. The engine and static-CD variants receive **fresh α, β draws from the Session 1 bootstrap on every path** plus a 5 bps per-trade cost, so the synthetic distribution reflects realistic parameter uncertainty rather than the self-consistent perfect-knowledge bound.

First, we generate the ensemble and run all seven strategies.

> __What should we see?__
>
> The scenario shape should be `(n_paths, T_total)` for market paths and `(n_paths, T_total, N)` for ticker prices. The noisy engine median $W/W_0$ should sit modestly above the static-CD and buy-and-hold baselines, not dramatically above: the 2014–2024 α/β point estimates are informative but noisy, and the engine's edge over static competitors is compressed once the allocator reads each path's noisy draw instead of the ground-truth generator params. Static-CD variants should stay near 1.0 (no dynamic reallocation), risk-free near 1.045.

In the code block below, we call [the `generate_hybrid_scenario(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.generate_hybrid_scenario), our noise-injected engine and static-CD wrappers, [the `backtest_buyhold(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.backtest_buyhold), and [the `backtest_buyhold_market(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.backtest_buyhold_market), and bind the following top-level variables:

| Binding | Type | Produced by |
|---|---|---|
| `scenario` | [`MyBacktestScenario`](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.MyBacktestScenario) | `generate_hybrid_scenario` |
| `result_eng` | [`MyBacktestResult`](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session3/#eCornellAIFinance.MyBacktestResult) | Noisy Cobb-Douglas engine (DD $\leq$ 15%, $\tau \leq$ 50%, $\alpha/\beta$ noisy, 5 bps) |
| `result_scd_warmup` | `MyBacktestResult` | Static CD with warmup-averaged $\lambda$, $\alpha/\beta$ noisy, 5 bps |
| `result_scd_neutral` | `MyBacktestResult` | Static CD with $\lambda = 0$, $\alpha/\beta$ noisy, 5 bps |
| `result_mv` | `MyBacktestResult` | S1 min-var buy-and-hold (fixed weights; no α/β input) |
| `result_eq` | `MyBacktestResult` | Equal-weight buy-and-hold |
| `result_mkt` | `MyBacktestResult` | Market (SPY surrogate) buy-and-hold |
| `result_rf` | `MyBacktestResult` | Deterministic risk-free path |
| `result_eng_raw` | `MyBacktestResult` | Noisy engine **without** bias correction, retained for the decomposition diagnostic below |
| `result_scd_floor` | `MyBacktestResult` | Alias for `result_scd_neutral`; the decomposition cell uses this name as the rebalancing-alpha floor |
| `bias_pct_per_yr` | `Float64` | Per-year bias-correction drag in %/yr (equal to `engine_raw_median - ENGINE_PRIOR_CCGR_PCT`), applied by `apply_bias_correction` to produce `result_eng` |
| `n_paths`, `n_steps` | `Int` | Scenario dimensions |

```julia
(; scenario, result_eng, result_eng_raw, result_scd_warmup, result_scd_neutral,
   result_scd_floor, result_mv, result_eq, result_mkt, result_rf,
   bias_pct_per_yr, n_paths, n_steps) = let

    # --- Step 1: Derived dimensions ---
    n_paths = N_STRESS_PATHS;
    n_steps = offset + N_ACTIVE_TRADING_DAYS;     # 84 warmup + 252 active

    # --- Step 2: Load or generate the 5,000-path scenario ---
    cache_path  = joinpath(_PATH_TO_DATA, "stress-scenario-cache.jld2");
    calib_hash  = hash(calib);
    cache_valid = false;
    local scen;
    if isfile(cache_path)
        c = load_results(cache_path);
        if c["n_paths"]    == n_paths        && c["n_steps"] == n_steps &&
           c["seed"]       == SCENARIO_SEED  && c["tickers"] == my_tickers &&
           c["calib_hash"] == calib_hash
            cache_valid = true;
            scen = c["scenario"]::MyBacktestScenario;
            println("Loaded cached scenario from $(cache_path) ($(n_paths) paths, seed=$(SCENARIO_SEED))")
        end
    end
    if !cache_valid
        println("Generating $(n_paths)-path hybrid-SIM scenario ($(n_steps) days, seed=$(SCENARIO_SEED))...")
        scen = generate_hybrid_scenario(market_model, portfolio, calib, my_tickers;
            n_paths      = n_paths,
            n_steps      = n_steps,
            Δt           = Δt,
            start_prices = start_prices,
            label        = "S2 Monte Carlo ($(n_paths)p × $(n_steps)d)",
            seed         = SCENARIO_SEED);
        save_results(cache_path, Dict(
            "scenario"   => scen,
            "n_paths"    => n_paths,
            "n_steps"    => n_steps,
            "seed"       => SCENARIO_SEED,
            "tickers"    => my_tickers,
            "calib_hash" => calib_hash,
        ));
        println("  Cached scenario to $(cache_path)")
    end
    println("  price_paths:  $(size(scen.price_paths))")
    println("  market_paths: $(size(scen.market_paths))")

    rules_params = (max_drawdown = TRIGGER_MAX_DRAWDOWN, max_turnover = TRIGGER_MAX_TURNOVER);

    # --- Step 3: Engine raw (frozen allocator, daily rebalance, dynamic λ) ---
    println("\nRunning Cobb-Douglas engine (frozen allocator, $(COST_BPS) bps cost) across $(n_paths) paths...")
    r_eng_raw = backtest_engine_frozen(scen, my_tickers, sim_params, rules_params;
        B₀ = B₀, offset = offset, L_short = L_short, L_long = L_long, L_growth = L_growth, GAIN = GAIN,
        cost_bps = COST_BPS);
    r_eng_raw.strategy_label = "Cobb-Douglas Engine raw (DD≤$(Int(TRIGGER_MAX_DRAWDOWN*100))%, τ≤$(Int(TRIGGER_MAX_TURNOVER*100))%, $(COST_BPS) bps)";

    # --- Step 4: Bias correction (option A, location shift) ---
    # The synthetic engine's median CCGR is structurally inflated relative to
    # what the engine delivers in real markets, primarily because the SIM has
    # iid residuals and daily rebalancing harvests free mean-reversion alpha
    # (the diversification return; Booth & Fama 1992). We anchor the corrected
    # central tendency to a long-run prior CCGR for an active multi-asset
    # strategy (rf + reasonable active-premium net of frictions; Damodaran ERP,
    # Ibbotson long-run series) so the synthetic ensemble characterizes
    # percentile and tail structure around a credible "typical year" rather
    # than a SIM-inflated point. Real-2025 then lands as an above-median draw
    # inside that ensemble, not as the calibration target.
    eng_med_ccgr_raw = let
        gs = Float64[];
        for p ∈ 1:n_paths
            w = r_eng_raw.wealth_paths[:, p];
            push!(gs, log(w[end] / w[1]) / ((length(w) - 1) * Δt) * 100);
        end
        median(gs)
    end
    bias_pct_per_yr = eng_med_ccgr_raw - ENGINE_PRIOR_CCGR_PCT;
    println("  Engine raw median CCGR: $(round(eng_med_ccgr_raw, digits=2))%/yr")
    println("  Real-2025 anchor:       $(ENGINE_PRIOR_CCGR_PCT)%/yr")
    println("  Bias correction drag:   $(round(bias_pct_per_yr, digits=2))pp/yr")

    r_eng = apply_bias_correction(r_eng_raw, bias_pct_per_yr, Δt);
    r_eng.strategy_label = "Cobb-Douglas Engine corrected (anchor=$(ENGINE_PRIOR_CCGR_PCT)%/yr prior, $(COST_BPS) bps)";

    # --- Step 5: Static-CD baselines (frozen allocator, no per-path noise) ---
    println("Running Static CD (warmup-avg λ, $(COST_BPS) bps) across $(n_paths) paths...")
    r_scd_w = backtest_static_cd_frozen(scen, my_tickers, sim_params, :warmup_avg;
        B₀ = B₀, offset = offset, L_short = L_short, L_long = L_long, L_growth = L_growth, GAIN = GAIN,
        cost_bps = COST_BPS);

    println("Running Static CD (neutral λ=0, $(COST_BPS) bps) across $(n_paths) paths...")
    r_scd_n = backtest_static_cd_frozen(scen, my_tickers, sim_params, :neutral;
        B₀ = B₀, offset = offset, L_short = L_short, L_long = L_long, L_growth = L_growth, GAIN = GAIN,
        cost_bps = COST_BPS);

    # The static-CD-floor (λ=0, no rebalance) is the SIM-prior allocation with
    # no rebalancing alpha; it bounds the engine's value-add from below in the
    # decomposition cell.
    r_scd_floor = r_scd_n;

    # --- Step 6: Static baselines that do not consume α, β (weights are fixed inputs) ---
    eqw = fill(1.0 / N, N);
    r_mv  = backtest_buyhold(scen, my_tickers; B₀ = B₀, offset = offset, weights = allocation_weights);
    r_mv.strategy_label = "S1 Min-Var (buy-and-hold)";
    r_eq  = backtest_buyhold(scen, my_tickers; B₀ = B₀, offset = offset, weights = eqw);
    r_eq.strategy_label = "Equal-Weight (buy-and-hold)";
    r_mkt = backtest_buyhold_market(scen; B₀ = B₀, offset = offset);
    r_mkt.strategy_label = "Market (SPY surrogate)";

    # --- Step 7: Risk-free deterministic baseline ---
    T_active = N_ACTIVE_TRADING_DAYS * Δt;
    rf_terminal = B₀ * exp(g_f * T_active);
    r_rf = MyBacktestResult();
    r_rf.scenario_label = scen.label;
    r_rf.strategy_label = "Risk-Free $(round(g_f*100, digits=1))%";
    r_rf.final_wealth   = fill(rf_terminal, n_paths);
    r_rf.max_drawdowns  = zeros(n_paths);
    r_rf.sharpe_ratios  = zeros(n_paths);
    rf_T = size(r_mv.wealth_paths, 1);
    r_rf.wealth_paths = repeat([B₀ * exp(g_f * (t-1) * Δt) for t in 1:rf_T], 1, n_paths);

    # --- Step 8: Median-W/W₀ summary ---
    strategies = (r_eng, r_eng_raw, r_scd_w, r_scd_n, r_mv, r_eq, r_mkt, r_rf);
    summary_df = DataFrame(
        "Strategy"    => [r.strategy_label for r ∈ strategies],
        "Median W/W₀" => [round(median(r.final_wealth) / B₀, digits=3) for r ∈ strategies],
    );
    println("\nAll strategies run. Median final W/W₀ by strategy:")
    
    # print the summary table with a compact text format that fits in the display -
    pretty_table(summary_df; backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))

    # package up the results for return -
    (scenario = scen, result_eng = r_eng, result_eng_raw = r_eng_raw,
     result_scd_warmup = r_scd_w, result_scd_neutral = r_scd_n,
     result_scd_floor = r_scd_floor,
     result_mv = r_mv, result_eq = r_eq, result_mkt = r_mkt, result_rf = r_rf,
     bias_pct_per_yr = bias_pct_per_yr,
     n_paths = n_paths, n_steps = n_steps)
end;
```

### Decomposition: where does the synthetic engine return come from?

The raw synthetic engine's median CCGR is structurally inflated relative to what the engine delivers on real markets, so we first attribute the synthetic return to its components on the *same* paths to isolate the share that comes from a known SIM artifact:

| Component | Median CCGR | What it captures |
|---|---|---|
| Equal-weight buy-and-hold | $(~7-8\%)$ | Passive baseline; tracks the synthetic market |
| Static Cobb-Douglas (λ=0, no rebalance) | $(~2\%)$ | SIM-prior allocation alone, no rebalancing alpha |
| Engine raw (daily rebalance, dynamic λ) | $(~30\%)$ | Adds the SIM-implied diversification return |

The gap between Static CD and Engine raw, about 28 pp/yr, is *daily-rebalancing alpha extracted on iid SIM residuals* (the **diversification return**; Booth & Fama 1992, Bouchey et al. 2012). In real markets residuals have positive daily autocorrelation (~0.05 to 0.10) so most of this premium does not materialize: the realized 2025 engine extracted +9.5 pp above the SIM-prior static baseline rather than +28 pp, and *that* +9.5 pp is the engine's real allocation alpha.

> __Bias-correction anchor + how to read the corrected scorecard:__
>
> We anchor the corrected synthetic median CCGR to the `ENGINE_PRIOR_CCGR_PCT` hyperparameter via a uniform per-step exponential drag (`apply_bias_correction`); the cell below sweeps five plausible priors so we can substitute our own and read the conclusion accordingly. The qualitative result (engine wins on drawdowns, passive wins on median return) is stable across the plausible prior range; only the magnitude shifts. The corrected ensemble preserves the engine's per-path rank order, drawdown shape, and volatility envelope; only the level is calibrated.

___

### Hyperparameter sensitivity: engine vs passive at five prior values

In the code block below, we sweep the bias-correction anchor over the grid `PRIOR_SENSITIVITY_GRID` and report the engine's corrected median CCGR alongside the gap to S1 Min-Var (which is not bias-corrected; it sits at its synthetic-natural median). Read the row matching whatever prior the audience considers reasonable.

```julia
let
    # --- Step 1: Compute baseline medians (Min-Var and raw Engine) on the shared scenario ---
    rows = NamedTuple[];
    mv_med_ccgr = let
        gs = Float64[];
        for p ∈ 1:n_paths
            w = result_mv.wealth_paths[:, p];
            # Annualized continuously compounded growth rate (%/yr):
            # g = (1/T) · log(W_T / W_0), then scaled to percent.
            push!(gs, log(w[end] / w[1]) / ((length(w) - 1) * Δt) * 100);
        end
        median(gs)
    end;
    eng_raw_med_ccgr = let
        gs = Float64[];
        for p ∈ 1:n_paths
            w = result_eng_raw.wealth_paths[:, p];
            push!(gs, log(w[end] / w[1]) / ((length(w) - 1) * Δt) * 100);
        end
        median(gs)
    end;

    # --- Step 2: Sweep the bias-correction anchor over PRIOR_SENSITIVITY_GRID ---
    # For each prior, bias-correct the engine paths then compute per-path CCGR
    # and Sharpe. The engine median is anchored to `prior` by construction.
    for prior ∈ PRIOR_SENSITIVITY_GRID
        # Per-year drag (pp/yr) that pulls the raw engine median down to the prior.
        bias = eng_raw_med_ccgr - prior;
        r_eng_at_prior = apply_bias_correction(result_eng_raw, bias, Δt);
        eng_ccgr_paths = [log(r_eng_at_prior.wealth_paths[end, p] / r_eng_at_prior.wealth_paths[1, p]) /
                          ((size(r_eng_at_prior.wealth_paths, 1) - 1) * Δt) * 100 for p ∈ 1:n_paths];
        eng_sharpe_paths = let
            T_yr = (size(r_eng_at_prior.wealth_paths, 1) - 1) * Δt;
            ss = Float64[];
            for p ∈ 1:n_paths
                w = r_eng_at_prior.wealth_paths[:, p];
                # Practitioner annualized volatility (%/yr): std of daily log-returns × √252.
                v = std(diff(log.(w))) * sqrt(252) * 100;
                # Per-path annualized CCGR (same formula as above, scaled to %/yr).
                g = log(w[end] / w[1]) / T_yr * 100;
                # Sharpe = (excess growth) / (vol). Guard against deterministic paths with v ≈ 0.
                push!(ss, v > 1e-6 ? (g - g_f * 100) / v : 0.0);
            end
            ss
        end;
        push!(rows, (
            Prior_pct  = prior,
            Drag_pp    = round(bias, digits=2),
            Eng_med_g  = round(median(eng_ccgr_paths), digits=2),
            MV_med_g   = round(mv_med_ccgr,            digits=2),
            Eng_minus_MV_pp = round(median(eng_ccgr_paths) - mv_med_ccgr, digits=2),
            Eng_med_Sharpe  = round(median(eng_sharpe_paths), digits=3),
        ));
    end
    # --- Step 3: Render the sensitivity table ---
    df = DataFrame(rows);
    pretty_table(df; backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))
    println()
    println("Read the row matching the prior you consider plausible. Engine_minus_MV_pp")
    println("> 0 means the engine beats Min-Var on median CCGR at that prior; < 0 means")
    println("Min-Var wins on median (engine still wins on drawdown; see scorecard below).")
end;
```

With the per-path arrays in hand, we assemble the distributional scorecard with VaR, CVaR, drawdown, Sharpe, and NPV metrics for each strategy. The engine row uses the bias-corrected wealth paths (`result_eng`); the raw uncorrected engine is summarized in the decomposition cell above and is also retained as `result_eng_raw` for reference.

`Med_Sharpe` is recomputed per path from `wealth_paths` as continuously-compounded excess growth over annualized log-return volatility, so every row is in the same units regardless of which backtest produced it. The NPV columns apply a uniform discount factor $e^{-g_f \cdot T}$ across strategies, so the risk-free row has $\text{NPV} \equiv 0$ by construction and serves as the zero reference.

In the code block below, we build the scorecard `DataFrame` and return `scorecard::DataFrame`.

```julia
scorecard = let
    # --- Step 1: Discount factor for the scorecard ---
    T_active      = N_ACTIVE_TRADING_DAYS * Δt;   # horizon in years
    discount      = exp(-g_f * T_active);          # NPV discount factor
    rf_growth_pct = g_f * 100;                     # risk-free CC growth rate (%/yr)

    # --- Step 2: Build rows ---
    rows = [
        tail_metrics(r, B₀, discount, Δt, rf_growth_pct)
        for r ∈ (result_eng, result_scd_warmup, result_scd_neutral,
                 result_mv, result_eq, result_mkt, result_rf)
    ];
    df = DataFrame(rows);

    pretty_table(df; backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))

    df
end;
```

Before drawing inferences from the scorecard, we sanity-check Monte Carlo convergence on the heaviest column. The 5% CVaR is the most sample-hungry metric because it averages over only the worst 5% of paths; on this run that is 5% of `n_paths`. The rule of thumb is that the standard error of the CVaR estimate should be small relative to the CVaR itself; if any non-deterministic ratio exceeds ~1%, we increase the path count.

In the code block below, we build a small convergence `DataFrame` and display it via `pretty_table`.

```julia
let
    # --- Step 1: Compute SE / |CVaR| percent for every scorecard row ---
    # A large ratio (above roughly 1%) signals the sample size is too small
    # to trust the CVaR estimate; the risk-free row is deterministic (ratio = 0).
    n_tail = floor(Int, 0.05 * n_paths);
    rows = NamedTuple[];
    for row ∈ eachrow(scorecard)
        ratio_pct = (row.CVaR5_W_W0 > 0 && row.CVaR5_SE_W_W0 > 0) ?
                    round(row.CVaR5_SE_W_W0 / abs(row.CVaR5_W_W0) * 100, digits=2) :
                    0.0;
        push!(rows, (
            Portfolio       = row.Portfolio,
            CVaR5_W_W0      = row.CVaR5_W_W0,
            CVaR5_SE_W_W0   = row.CVaR5_SE_W_W0,
            SE_pct_CVaR     = ratio_pct,
        ));
    end
    # --- Step 2: Render the ratio table ---
    cdf = DataFrame(rows);
    pretty_table(cdf; backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))
end;
```

### How do we compare against the real calendar-2025 plus 2026 YTD path?

The code block below overlays the **real calendar-2025 plus 2026 year-to-date trajectory** as a darkorange line on top of both cone panels. The engine cone uses the bias-corrected ensemble; the S1 min-var cone is the raw synthetic ensemble (passive strategies are not bias-corrected because the SIM rebalancing-alpha artifact only affects daily-rebalanced strategies). The orange line is the same engine and S1 min-var run on the real OHLC for the extended 326-trading-day window ([the `MyExtendedTestingMarketDataSet(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session1/#eCornellAIFinance.MyExtendedTestingMarketDataSet), 2025-01-02 through 2026-04-22).

> __What should we see?__
>
> Because the engine cone is anchored to the long-run prior CCGR, and 2025 plus 2026 YTD have been an above-typical period for US large-cap, the real orange line should land *above* the synthetic median for the engine and well above for S1 min-var. The shape of the orange path relative to the corrected band is the diagnostic: the real engine's tightness in drawdown (~3% in 2025, similar early 2026) keeps it well inside the 68% band, while the realized return puts it at the upper end of the cone. Because real horizon ($T_{\text{real}}$ ≈ 326 trading days) exceeds the synthetic horizon ($N_{\text{ACTIVE\_TRADING\_DAYS}}$ = 252), the orange line continues past the right edge of the synthetic cones; we extend the x-axis to accommodate.

The cell binds `p_mv_real::Plots.Plot` and `p_eng_real::Plots.Plot`, the two cone-panel plots (min-var and engine) that Task 1's scorecard cells render.

```julia
p_mv_real, p_eng_real = let
    # --- Step 1: Load real 2025 + 2026 YTD daily OHLC for the same universe ---
    ds = MyExtendedTestingMarketDataSet()["dataset"]::Dict{String,DataFrame};
    haskey(ds, "SPY") || error("SPY missing from MyExtendedTestingMarketDataSet.");
    mkt_r = Float64.(ds["SPY"].close);
    T_r   = length(mkt_r);

    pmat_r = zeros(T_r, N + 1);
    for t ∈ 1:T_r
        pmat_r[t, 1] = t;
    end
    for (k, ticker) ∈ enumerate(my_tickers)
        haskey(ds, ticker) || error("$(ticker) missing from real dataset.");
        pmat_r[:, k + 1] = Float64.(ds[ticker].close);
    end

    # --- Step 2: Sentiment signal and smoothed market growth from real SPY ---
    ema_s_r  = compute_ema(mkt_r; window = L_short);
    ema_l_r  = compute_ema(mkt_r; window = L_long);
    λ_r      = compute_lambda(ema_s_r, ema_l_r; G = GAIN);
    gm_raw_r = compute_market_growth(mkt_r; Δt = Δt);
    gm_ema_r = compute_ema(gm_raw_r; window = L_growth);

    # --- Step 3: Engine run on real extended window with same triggers + cost as synthetic ---
    ctx_r = build(MyRebalancingContextModel, (
        B = B₀, tickers = my_tickers, marketdata = pmat_r,
        marketfactor = gm_ema_r, sim_parameters = sim_params,
        lambda = 0.0, Δt = Δt, epsilon = 0.1,
    ));
    n_trading_r = T_r - offset;
    rules_r = build(MyTriggerRules, (
        max_drawdown = TRIGGER_MAX_DRAWDOWN, max_turnover = TRIGGER_MAX_TURNOVER,
        rebalance_schedule = ones(Int, n_trading_r),
    ));
    eng_results_r = run_rebalancing_engine(ctx_r, rules_r, λ_r;
        offset = offset, allocator = :cobb_douglas, cost_bps = COST_BPS);
    eng_wealth_r = compute_wealth_series(eng_results_r, pmat_r, my_tickers; offset = offset);

    # --- Step 4: S1 min-var buy-and-hold on extended real window ---
    buyin_r     = offset + 1;
    p_start_r   = [pmat_r[buyin_r, k + 1] for k ∈ 1:N];
    shares_mv_r = B₀ .* allocation_weights ./ p_start_r;
    mv_wealth_r = vcat([B₀], [sum(shares_mv_r[k] * pmat_r[t, k + 1] for k ∈ 1:N) for t ∈ buyin_r:T_r]);

    length(eng_wealth_r) == length(mv_wealth_r) ||
        error("engine ($(length(eng_wealth_r))) and min-var ($(length(mv_wealth_r))) wealth lengths disagree");

    eng_r = eng_wealth_r ./ B₀;
    mv_r  = mv_wealth_r  ./ B₀;

    # --- Step 5: CCGR-percentile diagnostic (W/W₀ comparison is apples-to-oranges
    #             because real-extended has $(n_trading_r) active days vs $(N_ACTIVE_TRADING_DAYS) synthetic) ---
    yr_real = n_trading_r * Δt;
    yr_syn  = N_ACTIVE_TRADING_DAYS * Δt;
    eng_ccgr_real = log(eng_r[end] / eng_r[1]) / yr_real * 100;
    mv_ccgr_real  = log(mv_r[end]  / mv_r[1])  / yr_real * 100;
    eng_ccgr_syn  = [log(result_eng.wealth_paths[end, p] / result_eng.wealth_paths[1, p]) / yr_syn * 100 for p ∈ 1:n_paths];
    mv_ccgr_syn   = [log(result_mv.wealth_paths[end, p]  / result_mv.wealth_paths[1, p])  / yr_syn * 100 for p ∈ 1:n_paths];
    eng_pct = round(mean(eng_ccgr_syn .< eng_ccgr_real) * 100, digits=1);
    mv_pct  = round(mean(mv_ccgr_syn  .< mv_ccgr_real)  * 100, digits=1);

    # --- Step 6: Wide-form CCGR comparison DataFrame (rows = strategy) ---
    ccgr_df = DataFrame(
        "Strategy"                       => ["Engine", "S1 Min-Var"],
        "Real CCGR (%/yr)"               => [
            round(eng_ccgr_real, digits = 2),
            round(mv_ccgr_real,  digits = 2),
        ],
        "Synthetic median (%/yr)"        => [
            round(median(eng_ccgr_syn), digits = 2),
            round(median(mv_ccgr_syn),  digits = 2),
        ],
        "Synthetic P5 (%/yr)"            => [
            round(quantile(eng_ccgr_syn, 0.05), digits = 2),
            round(quantile(mv_ccgr_syn,  0.05), digits = 2),
        ],
        "Synthetic P95 (%/yr)"           => [
            round(quantile(eng_ccgr_syn, 0.95), digits = 2),
            round(quantile(mv_ccgr_syn,  0.95), digits = 2),
        ],
        "Real percentile in synthetic"   => ["P$(eng_pct)", "P$(mv_pct)"],
    );
    println("Real 2025 + 2026 YTD CCGR vs corrected synthetic ensemble ($(n_trading_r) real active days, $(N_ACTIVE_TRADING_DAYS) synthetic, both annualized):")
    pretty_table(ccgr_df; backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))

    # --- Step 7: Build the two overlay panels ---
    n_active = N_ACTIVE_TRADING_DAYS;
    n_tail   = max(1, floor(Int, 0.05 * n_paths));
    worst_mv_idx  = partialsortperm(result_mv.final_wealth,  1:n_tail);
    worst_eng_idx = partialsortperm(result_eng.final_wealth, 1:n_tail);
    rf_multiple   = result_rf.final_wealth[1] / B₀;

    mv_w_syn  = result_mv.wealth_paths  ./ B₀;
    eng_w_syn = result_eng.wealth_paths ./ B₀;
    mv_c      = cones(mv_w_syn);
    eng_c     = cones(eng_w_syn);
    days_syn  = 0:n_active;
    days_real = 0:(length(eng_r) - 1);

    all_lo = min(minimum(mv_c.p005), minimum(eng_c.p005), minimum(mv_r), minimum(eng_r)) * 0.98;
    all_hi = max(maximum(mv_c.p995), maximum(eng_c.p995), maximum(mv_r), maximum(eng_r)) * 1.02;

    # x-axis extends to whichever of (synthetic horizon, real horizon) is longer
    xmax = max(maximum(days_syn), maximum(days_real));

    p_mv = panel(mv_w_syn, mv_c, worst_mv_idx,
        "S1 Min-Var: synthetic cones with real 2025 + 2026 YTD overlay",
        days_syn, n_tail, rf_multiple; show_ylabel = true);
    plot!(p_mv, days_real, mv_r;
        lw = 4, c = :darkorange, label = "Real 2025 + 2026 YTD (S1 min-var)");
    plot!(p_mv, ylims = (all_lo, all_hi), xlims = (0, xmax));

    p_eng = panel(eng_w_syn, eng_c, worst_eng_idx,
        "Cobb-Douglas Engine: corrected synthetic cones with real 2025 + 2026 YTD overlay",
        days_syn, n_tail, rf_multiple; show_ylabel = false);
    plot!(p_eng, days_real, eng_r;
        lw = 4, c = :darkorange, label = "Real 2025 + 2026 YTD (engine)");
    plot!(p_eng, ylims = (all_lo, all_hi), xlims = (0, xmax));

    (p_mv, p_eng)
end;
```

### Hand-off: saving the Monte Carlo arrays for downstream sessions
We save the per-path arrays (final wealth, max drawdowns, Sharpe ratios, NPV) for each strategy and the per-path market-return array so Session 3 / Session 4 examples can consume them without re-running the multi-path backtest. The file is written to `stress-test-engine.jld2`, the canonical Session 2 hand-off.

```julia
let
    T_active = N_ACTIVE_TRADING_DAYS * Δt;
    discount = exp(-g_f * T_active);
    buyin_day = offset + 1;

    # --- Step 1: Per-path market stats from the underlying scenario ---
    # Per-path market-stat snapshots for downstream Session 3 / Session 4
    # regime-conditional analysis (saves the next session re-running this 5,000-path
    # scenario just to recover the per-path market return).
    market_terminal = [scenario.market_paths[p, end] for p ∈ 1:n_paths];
    market_start    = [scenario.market_paths[p, buyin_day] for p ∈ 1:n_paths];  # buy-in day = offset + 1
    market_return   = (market_terminal ./ market_start) .- 1.0;                 # simple return over active window

    save_path = joinpath(_PATH_TO_DATA, "stress-test-engine.jld2");
    save_results(save_path, Dict(
        "n_paths"             => n_paths,
        "n_steps"             => n_steps,
        "offset"              => offset,
        "g_f"                 => g_f,
        "B₀"                  => B₀,
        "discount"            => discount,
        "my_tickers"          => my_tickers,
        "allocation_weights"  => allocation_weights,
        # Per-path market stats (for regime-conditional analysis)
        "market_terminal"     => market_terminal,
        "market_start"        => market_start,
        "market_return"       => market_return,
        # Engine — RAW arrays (downstream notebooks expect raw under eng_* for
        # paired comparisons against their own raw engines: TurnoverAttribution,
        # RegimeAwareSentiment, Maya intro).
        "eng_final_wealth"       => result_eng_raw.final_wealth,
        "eng_max_drawdowns"      => result_eng_raw.max_drawdowns,
        "eng_sharpe_ratios"      => result_eng_raw.sharpe_ratios,
        "eng_npv"                => result_eng_raw.final_wealth .* discount .- B₀,
        # Engine — bias-corrected arrays (anchored to real-2025 CCGR via
        # apply_bias_correction). Opt-in for consumers that want the corrected
        # basis; bias_pct_per_yr stored alongside for reproducibility.
        "eng_corr_final_wealth"  => result_eng.final_wealth,
        "eng_corr_max_drawdowns" => result_eng.max_drawdowns,
        "eng_corr_sharpe_ratios" => result_eng.sharpe_ratios,
        "eng_corr_npv"           => result_eng.final_wealth .* discount .- B₀,
        "bias_pct_per_yr"        => bias_pct_per_yr,
        # Static Cobb-Douglas, warmup-avg λ
        "scd_w_final_wealth"  => result_scd_warmup.final_wealth,
        "scd_w_max_drawdowns" => result_scd_warmup.max_drawdowns,
        "scd_w_sharpe_ratios" => result_scd_warmup.sharpe_ratios,
        "scd_w_npv"           => result_scd_warmup.final_wealth .* discount .- B₀,
        # Static Cobb-Douglas, neutral λ=0
        "scd_n_final_wealth"  => result_scd_neutral.final_wealth,
        "scd_n_max_drawdowns" => result_scd_neutral.max_drawdowns,
        "scd_n_sharpe_ratios" => result_scd_neutral.sharpe_ratios,
        "scd_n_npv"           => result_scd_neutral.final_wealth .* discount .- B₀,
        # S1 min-var
        "mv_final_wealth"     => result_mv.final_wealth,
        "mv_max_drawdowns"    => result_mv.max_drawdowns,
        "mv_sharpe_ratios"    => result_mv.sharpe_ratios,
        "mv_npv"              => result_mv.final_wealth .* discount .- B₀,
        # Equal-weight
        "eq_final_wealth"     => result_eq.final_wealth,
        "eq_max_drawdowns"    => result_eq.max_drawdowns,
        "eq_sharpe_ratios"    => result_eq.sharpe_ratios,
        "eq_npv"              => result_eq.final_wealth .* discount .- B₀,
        # Market (SPY surrogate)
        "mkt_final_wealth"    => result_mkt.final_wealth,
        "mkt_max_drawdowns"   => result_mkt.max_drawdowns,
        "mkt_sharpe_ratios"   => result_mkt.sharpe_ratios,
        "mkt_npv"             => result_mkt.final_wealth .* discount .- B₀,
        # Risk-free
        "rf_final_wealth"     => result_rf.final_wealth,
        "scenario_label"      => scenario.label,
    ));
    println("Saved Monte Carlo arrays to: $(save_path)")
    println("  Keys: engine / S1 min-var / equal-weight / market / risk-free")
    println("  Plus per-path market_return for downstream regime-conditional analysis")
end;
```

## Task 2: Engine on a Tech-Heavy Basket Through the 2025 Tariff Correction
In this task, we deploy the engine on January 2 2025 against a tech-heavy basket (NVDA, AVGO, AMD, MSFT, AAPL, AMZN, TSLA, NFLX, ADBE, CRM, ORCL, INTC, MU, QCOM) via [the `MyDeploymentMarketDataSet(...)` function](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session1/#eCornellAIFinance.MyDeploymentMarketDataSet) so the active window covers the entire 2025 + 2026 YTD path, including the **April 8 2025 tariff-correction low** (_liberation day_) where SPY fell 15% from buyin before recovering to a 21.65% gain by April 22 2026.

> __What this run tests, and what it does not:__
>
> Does the engine extract value on a basket with real α dispersion (the defensive S1 basket gave it little to concentrate into), and does the 15% drawdown trigger save capital during a real correction (the April 8 2025 low is exactly the event the trigger is designed for). Caveat: the 2025 + 2026 YTD period had AI/semiconductor leadership that aligned with the engine's calibrated alpha reads (NVDA at α = 0.35, AVGO at α = 0.19), and the post-correction recovery concentrated in those same names; a different regime (value rotation, defensive cycle) would not necessarily reproduce this result. 
> 
> This task demonstrates the engine's *mechanism* (daily concentration into highest-γ assets, drawdown trigger as risk circuit-breaker) on a real stress event, not a forward forecast.

In the code block below, we build the tech basket, pull SIM parameters from the S1 calibration, run the engine + equal-weight tech + SPY on the deployment-window real data (Jan 2 2025 through April 22 2026 = 326 active trading days = 1.29 years), and report a wealth-and-risk scorecard. The cell returns:

* `tech_tickers::Vector{String}`: tech-basket ticker symbols (NVDA, AVGO, AMD, MSFT, AAPL, and peers).
* `sim_params_tech::Dict{String,Tuple{Float64,Float64,Float64}}`: SIM adapter dict (α, β, σ_ε per ticker) built from the Session 1 calibration, consumed by the noisy-engine wrapper.
* `eng_real_tech::Vector{Float64}`: engine wealth trajectory on the real deployment window.
* `eq_real_tech::Vector{Float64}`: equal-weight-tech wealth trajectory on the same window.
* `spy_real_tech::Vector{Float64}`: SPY buy-and-hold wealth trajectory (market benchmark).
* `eng_results_tech::Dict{Int,MyRebalancingResult}`: per-day engine results, consumed by Task 3 for the trade diagnostic.
* `n_trading_tech::Int`: active-window trading-day count (326, equal to the real window length minus the 84-day EMA warmup offset).
* `pmat_real_tech::Matrix{Float64}`: real tech-basket price matrix for the active window (day index in column 1, per-ticker close in columns 2..N+1).

```julia
(; tech_tickers, sim_params_tech, eng_real_tech, eq_real_tech, spy_real_tech,
   eng_results_tech, n_trading_tech, pmat_real_tech) = let

    # --- Step 1: Build tech-heavy basket from S1 calibration + portfolio surrogate ---
    candidate = ["AAPL","MSFT","NVDA","META","GOOGL","AMZN","TSLA","AMD","NFLX",
                 "ADBE","CRM","ORCL","AVGO","INTC","MU","QCOM"];
    snap_lookup = Dict(MyCurrentPrices()["tickers"] .=> MyCurrentPrices()["prices"]);
    in_calib = filter(t -> t ∈ calib["tickers"], candidate);
    in_surr  = filter(t -> haskey(MyPortfolioSurrogateModel()["marginals"], t), in_calib);
    tech_tk  = filter(t -> haskey(snap_lookup, t), in_surr);
    Nt = length(tech_tk);
    println("Tech-heavy basket: $(Nt) tickers = $(tech_tk)")

    # --- Step 2: Build sim_params (α, β, σ_ε) for tech basket ---
    sp_tech = Dict{String,Tuple{Float64,Float64,Float64}}();
    for t ∈ tech_tk
        i = findfirst(==(t), calib["tickers"]);
        sp_tech[t] = (calib["alpha"][i], calib["beta"][i], calib["sigma_eps"][i]);
    end

    # --- Step 3: Load extended real data and build pmat ---
    ds_real = MyDeploymentMarketDataSet()["dataset"]::Dict{String,DataFrame};
    haskey(ds_real, "SPY") || error("SPY missing from deployment dataset.");
    mkt_r = Float64.(ds_real["SPY"].close);
    T_r   = length(mkt_r);
    n_trading_r = T_r - offset;
    pmat_r = zeros(T_r, Nt + 1);
    for t ∈ 1:T_r; pmat_r[t, 1] = t; end
    for (k, ticker) ∈ enumerate(tech_tk)
        haskey(ds_real, ticker) || error("$(ticker) missing from real data.");
        pmat_r[:, k + 1] = Float64.(ds_real[ticker].close);
    end

    # --- Step 4: Sentiment + smoothed market growth from real SPY ---
    ema_s_r  = compute_ema(mkt_r; window = L_short);
    ema_l_r  = compute_ema(mkt_r; window = L_long);
    λ_r      = compute_lambda(ema_s_r, ema_l_r; G = GAIN);
    gm_raw_r = compute_market_growth(mkt_r; Δt = Δt);
    gm_ema_r = compute_ema(gm_raw_r; window = L_growth);

    # --- Step 5: Run engine on real tech basket (daily rebalance, 5 bps cost) ---
    ctx_t = build(MyRebalancingContextModel, (
        B = B₀, tickers = tech_tk, marketdata = pmat_r,
        marketfactor = gm_ema_r, sim_parameters = sp_tech,
        lambda = 0.0, Δt = Δt, epsilon = 0.1,
    ));
    rules_t = build(MyTriggerRules, (
        max_drawdown = TRIGGER_MAX_DRAWDOWN, max_turnover = TRIGGER_MAX_TURNOVER,
        rebalance_schedule = ones(Int, n_trading_r),
    ));
    eng_res_t = run_rebalancing_engine(ctx_t, rules_t, λ_r;
        offset = offset, allocator = :cobb_douglas, cost_bps = COST_BPS);
    eng_w_t = compute_wealth_series(eng_res_t, pmat_r, tech_tk; offset = offset);

    # --- Step 6: Equal-weight buy-and-hold on real tech ---
    buyin_r   = offset + 1;
    p_start_r = [pmat_r[buyin_r, k + 1] for k ∈ 1:Nt];
    shares_eq = (B₀ / Nt) ./ p_start_r;
    eq_w_t    = vcat([B₀], [sum(shares_eq[k] * pmat_r[t, k + 1] for k ∈ 1:Nt) for t ∈ buyin_r:T_r]);

    # --- Step 7: SPY single-asset buy-and-hold ---
    spy_shares = B₀ / mkt_r[buyin_r];
    spy_w_t    = vcat([B₀], [spy_shares * mkt_r[t] for t ∈ buyin_r:T_r]);

    # --- Step 8: Scorecard ---
    function realmetrics(label, w)
        horizon = (length(w) - 1) * Δt;
        g_pct   = log(w[end] / w[1]) / horizon * 100;
        log_rets = diff(log.(w));
        vol_pct  = std(log_rets) * sqrt(252) * 100;
        sharpe   = vol_pct > 1e-6 ? (g_pct - g_f * 100) / vol_pct : 0.0;
        peak     = accumulate(max, w);
        max_dd   = maximum((peak .- w) ./ peak) * 100;
        return (
            Strategy   = label,
            CCGR_pct   = round(g_pct,   digits=2),
            Vol_pct    = round(vol_pct, digits=2),
            Sharpe     = round(sharpe,  digits=3),
            Max_DD_pct = round(max_dd,  digits=2),
            W_over_W0  = round(w[end] / B₀, digits=3),
        );
    end
    rows = [
        realmetrics("Engine (cost=$(COST_BPS) bps, daily rebalance)", eng_w_t),
        realmetrics("Equal-Weight tech (no rebalance)",                eq_w_t),
        realmetrics("SPY (single asset, no rebalance)",                spy_w_t),
    ];
    df = DataFrame(rows);
    println("\nReal extended window: $(n_trading_r) active trading days = $(round(n_trading_r * Δt, digits=2)) years\n")
    pretty_table(df; backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))

    (tech_tickers = tech_tk, sim_params_tech = sp_tech,
     eng_real_tech = eng_w_t, eq_real_tech = eq_w_t, spy_real_tech = spy_w_t,
     eng_results_tech = eng_res_t, n_trading_tech = n_trading_r,
     pmat_real_tech = pmat_r)
end;
```

In the code block below, we plot the three real-data wealth trajectories on the same axis so the engine's relative performance is unambiguous.

```julia
let
    days = 0:n_trading_tech;
    p = plot(days, eng_real_tech ./ B₀;
        lw = 4, c = :steelblue, label = "Engine (Cobb-Douglas, daily rebalance)",
        xlabel = "Trading Day (after 84-day warmup)",
        ylabel = "Scaled Wealth (W_t / W₀)",
        title = "Tech-heavy basket, real 2025 + 2026 YTD: engine vs passive baselines",
        fontsize = 16, size = (1100, 600),
        left_margin = 12Plots.mm, right_margin = 6Plots.mm,
        top_margin = 6Plots.mm, bottom_margin = 8Plots.mm);
    plot!(p, days, eq_real_tech ./ B₀;  lw = 3, c = :firebrick, label = "Equal-Weight tech (no rebalance)");
    plot!(p, days, spy_real_tech ./ B₀; lw = 3, c = :seagreen,  label = "SPY (single asset, no rebalance)");
    hline!(p, [1.0]; lw = 1, ls = :dash, c = :black, alpha = 0.6, label = "Initial wealth (1.0)");
    plot!(p, bg = "gray95", background_color_outside = "white",
        framestyle = :box, fg_legend = :transparent, legend = :topleft);
    p
end
```

## Task 3: Daily Trade Diagnostic, What is the Engine Actually Doing?
In this task, we make the mechanics of _daily rebalancing_ concrete by inspecting the engine's per-day output from the Task 2 real-tech deployment, extracting dollar turnover, transaction cost, cash fraction, trigger fires, and top-3 concentration from the engine's `Dict{Int,MyRebalancingResult}` via the notebook-local `trade_diagnostics(...)` helper (defined in the Implementation section above).

> __What should we see?__
>
> **Day-0 (warmup-end, Dec 31 2024) cash fraction is high** because the late-2024 EMA-crossover signal was bearish (λ < 0) and smoothed market growth was negative; with all γ at the floor the engine starts defensively in cash, not fully invested. Through the early-2025 tariff correction the cash fraction modulates as λ swings, keeping the realized max drawdown at ~8.5% even though SPY drew down 19% over the same window; the 15% trigger never has to fire because the daily allocator's defensive posture absorbs the shock first. Daily turnover stays bounded by the $\tau_{\max}$ cap rather than swinging floor-to-full like the raw [`BuildCobbDouglasAllocator`](eCornell-AI-Finance-S2-Example-Core-BuildCobbDouglasAllocator-May-2026.ipynb) share plots, top-3 concentration sits moderate-to-high in the high-α leaders (NVDA, AVGO, AMD, MSFT), and total transaction cost lands in single-digit percent of starting capital at 5 bps × ~300 days × portfolio-fraction turnover.

In the cells below, we first run `trade_diagnostics` and unpack four bindings, then render two tables in turn: a window-level summary, and a per-day timeline at sampled points. The compute cell returns `diag::NamedTuple` (per-day engine mechanics), `max_dd_day::Int` (active-day index of the worst drawdown), `avg_turnover_frac::Float64` (mean daily dollar value traded as a fraction of $B_0$), and `dd_series::Vector{Float64}` (the running drawdown fraction along the engine's wealth path).

```julia
(; diag, max_dd_day, avg_turnover_frac, dd_series) = let
    # --- Step 1: Call the trade-diagnostics helper for per-day engine mechanics ---
    diag = trade_diagnostics(eng_results_tech, pmat_real_tech, tech_tickers;
        offset = offset, cost_bps = COST_BPS);

    # --- Step 2: Derive turnover fraction and locate the max-drawdown day ---
    avg_turnover_frac = mean(diag.daily_traded) / B₀;

    # Find max-drawdown day on the engine's wealth trajectory (post-warmup).
    # Running high-water mark: peak[t] = max(W_0, W_1, ..., W_t).
    peak = accumulate(max, diag.total_wealth);
    # Drawdown fraction: (peak - current) / peak, in [0, 1).
    dd_series = (peak .- diag.total_wealth) ./ peak;
    max_dd_day = argmax(dd_series) - 1;  # day index in 0..T

    (diag = diag, max_dd_day = max_dd_day,
     avg_turnover_frac = avg_turnover_frac, dd_series = dd_series)
end;
```

The code block below renders the **window-level summary**: active-trading-day count, cash-fraction posture (Day-0, mean, max), notional traded and transaction cost as fractions of $B_0$, daily turnover, trigger-fire count, max-drawdown day and magnitude, and top-3 concentration statistics.

```julia
let
    # --- Step 1: Build summary rows (one per metric) ---
    rows = [
        (Statistic = "Active trading days",                            Value = string(diag.T)),
        (Statistic = "Day-0 (warmup-end) shares value (× W₀)",         Value = string(round(diag.initial_trade / B₀, digits=4))),
        (Statistic = "Day-0 cash fraction",                            Value = string(round(diag.cash_frac[1] * 100, digits=2)) * "%"),
        (Statistic = "Mean cash fraction over window",                 Value = string(round(mean(diag.cash_frac) * 100, digits=2)) * "%"),
        (Statistic = "Max cash fraction over window",                  Value = string(round(maximum(diag.cash_frac) * 100, digits=2)) * "%"),
        (Statistic = "Total notional traded over window (× W₀)",       Value = string(round(sum(diag.daily_traded) / B₀, digits=2))),
        (Statistic = "Total transaction cost over window (× W₀)",      Value = string(round((diag.initial_cost + sum(diag.daily_cost)) / B₀, digits=6))),
        (Statistic = "Mean daily turnover (fraction of B₀)",           Value = string(round(avg_turnover_frac * 100, digits=2)) * "%"),
        (Statistic = "Trigger-fire days (drawdown > $(Int(TRIGGER_MAX_DRAWDOWN*100))%)", Value = string(sum(diag.triggered))),
        (Statistic = "Max-drawdown day (active day index)",            Value = string(max_dd_day)),
        (Statistic = "Max drawdown realized (%)",                      Value = string(round(maximum(dd_series) * 100, digits=2)) * "%"),
        (Statistic = "Mean top-3 concentration (fraction of portfolio)", Value = string(round(mean(diag.top3_concen) * 100, digits=2)) * "%"),
        (Statistic = "Max top-3 concentration over window",            Value = string(round(maximum(diag.top3_concen) * 100, digits=2)) * "%"),
    ];

    # --- Step 2: Render the summary table ---
    println("Engine trade-diagnostic summary (Task 2 tech-basket real-data run):\n")
    pretty_table(DataFrame(rows); backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))
end;
```

The code block below samples the engine at six characteristic days (warmup-end, max-drawdown day, quartiles, end-of-window) and reports the **per-day timeline**: top-3 tickers, top-3 concentration, cash fraction, dollar value traded as a fraction of $B_0$, and whether the drawdown trigger fired.

```julia
let
    # --- Step 1: Pick six characteristic days ---
    sample_days = sort(unique([0, max_dd_day, max(1, div(diag.T, 4)), div(diag.T, 2),
                               max(div(diag.T, 4) * 3, div(diag.T, 2) + 1), diag.T]));

    # --- Step 2: Build one row per sampled day ---
    rows = NamedTuple[];
    for d ∈ sample_days
        d_safe = clamp(d, 0, diag.T);
        t1, t2, t3 = diag.top3_tickers[d_safe + 1];
        traded = d_safe == 0 ? diag.initial_trade : diag.daily_traded[d_safe];
        triggered = d_safe == 0 ? false : diag.triggered[d_safe];
        push!(rows, (
            Day              = d_safe,
            Top1             = t1,
            Top2             = t2,
            Top3             = t3,
            Top3_concen_pct  = round(diag.top3_concen[d_safe + 1] * 100, digits=2),
            Cash_pct         = round(diag.cash_frac[d_safe + 1] * 100,  digits=2),
            Daily_traded_W0  = round(traded / B₀, digits=4),
            Trigger_fired    = triggered,
        ));
    end

    # --- Step 3: Render the timeline table ---
    println("Per-day timeline at sampled points (warmup-end, max-DD day, quartiles, end-of-window):\n")
    pretty_table(DataFrame(rows); backend = :text,
        fit_table_in_display_horizontally = false,
        fit_table_in_display_vertically = false,
        table_format = TextTableFormat(borders = text_table_borders__compact))
end;
```

## Summary
This example evaluated the Cobb-Douglas rebalancing engine across a Monte Carlo ensemble of generative SIM paths with realistic per-trade costs, anchored the synthetic edge to a tunable long-run prior CCGR via a bias-correction sensitivity sweep, and then validated the mechanism on a real tech-heavy deployment through the 2025 tariff correction with a per-day trade diagnostic.

> __Key Takeaways:__
>
> * __Most synthetic engine outperformance is a generator artifact:__ On simulated paths with iid residuals, daily rebalancing harvests a diversification return that does not exist in real markets. Bias-correct any forward claim against this artifact and report the corrected edge with a sensitivity band over plausible long-run priors.
> * __An adaptive allocator's edge is universe-dependent:__ The same engine can shift from a drawdown-discipline strategy on a defensive basket to a return-enhancement strategy on a high-α basket, depending on the alignment between calibrated parameters and the deployment regime. Universe selection is a strategic decision, not a bookkeeping detail, and motivates the online estimation that follows in Session 3.
> * __Daily rebalancing is a mechanical activity with measurable footprints:__ A trade-level diagnostic exposes how often the allocator trades, what it pays in friction, and how concentrated its positions are. These mechanics, plus the drawdown trigger that operationalizes risk discipline, are the inputs a risk committee needs to sign off on a live deployment.

The construction used here (regime-switching market factor, Student-t copula across the cross-section, per-ticker generative residuals) mirrors the stress-testing pattern major capital-markets firms have deployed in production.

> __Industry analog (anonymized case study):__
>
> A US capital-markets firm reported using generative models to simulate multi-factor stress paths, produce realistic non-linear correlations, and enable faster Monte Carlo scenario generation. The deployment cut scenario-generation time from hours to minutes, achieved tail-event coverage validated against historical crises, and was integrated into firmwide ICAAP capital planning. 
> 
> _What did this notebook teach?_ This notebook teaches the methodology behind that pattern: the JumpHMM regime model carries non-Gaussian aggregate paths, the Student-t copula preserves cross-sectional dependence under tail events, and the SIM construction scales per-ticker stress with each ticker's calibrated $\sigma_{\varepsilon}$. The bias-correction discussion above is the honest qualifier on going from a fast scenario engine to a decision-grade signal.
