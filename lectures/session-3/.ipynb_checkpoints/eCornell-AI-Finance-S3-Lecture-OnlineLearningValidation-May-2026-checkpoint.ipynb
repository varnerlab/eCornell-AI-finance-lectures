# Session 3: Teaching the Engine to Learn

In Session 2, we built an adaptive rebalancing engine that reads market sentiment and adjusts allocation in real time. The engine works, but three of its inputs are still incomplete: the SIM parameters $(\alpha_i, \beta_i, \sigma_{\varepsilon,i})$ were calibrated once from historical data, the universe of tradeable tickers was hand-picked once from the Session 1 configuration, and the news-aware preference weight declared in Session 2 had no ingestion pipeline behind it.

__Challenge:__ All three of these inputs feed the same downstream object, the Cobb-Douglas preference weight $\gamma_i$, so the engine's allocation is only as good as how we update them. In this session, we replace frozen SIM parameters with online estimation, replace the static universe with a learning algorithm that picks the basket adaptively, and stand up the LLM-driven news pipeline that fills the news factor. We then validate the result against formal deployment gates. Three classic AI techniques run the chain.
* **LLM-driven signal extraction.** Each headline is routed to Claude, which returns a numeric sentiment score in $[-1, +1]$. Per-ticker daily aggregation produces the news factor that the news-aware Single Index Model regression consumes; the regression then recovers the per-ticker loading $\nu_i$ alongside the SIM coefficients.
* **Exponentially weighted least squares** is the streaming counterpart of batch OLS, mathematically equivalent to stochastic gradient descent with an exponentially decaying step size; it is the building block of every online-learning system from Kalman filters to adaptive signal processing. 
* **Multi-armed bandits** are the canonical reinforcement-learning problem: an agent repeatedly selects from a discrete action set, observes a stochastic reward, and trades off exploration against exploitation. The epsilon-greedy policy with decaying exploration achieves provably sublinear regret: the learner converges on the best arm while never stopping exploring entirely. 

All three are mature AI; their application to the news-factor, SIM-parameter, and universe-selection problems here is the session's contribution.

> __Learning Objectives:__
>
> By the end of this session, you will be able to:
> * __Frame portfolio allocation as a reinforcement-learning problem:__ State the agent / state / action / reward / policy quadruple for a daily-rebalance engine, and identify which RL family (multiplicative weights, bandits, Q-learning) fits which action space and feedback structure.
> * __Apply an epsilon-greedy multi-armed bandit:__ Implement the framework over a finite arm set with the standard $t^{-1/3}(K \ln t)^{1/3}$ exploration schedule, and explain why it converges on the best arm while never stopping exploration entirely.
> * __Use the bandit to learn a tradeable ticker universe:__ Apply it to the combinatorial $2^{K}-1$ subset action space with Cobb-Douglas utility as the reward, and read the paired-path verdict on whether bandit-selected universes earn their complexity against the hand-picked Session 1 baseline.
> * __Validate before deployment:__ Evaluate strategy performance against five pass/fail deployment gates (Sharpe, drawdown, failure rate, wealth versus the frozen baseline, median NPV) and export the compliance configuration that defines the pre-approved operating limits for Session 4's production system.

Let's teach the engine to learn.

## Rewind: What did we do in Session 2?
Instead of asking _what portfolio minimizes variance?_ we ask a different question: **given a budget and current market conditions, what portfolio maximizes the investor's utility (happiness)?**

* __Background__: Utility maximization did not start in finance. It has been the workhorse of consumer choice theory since [Daniel Bernoulli's 1738 treatment of the St. Petersburg paradox](https://en.wikipedia.org/wiki/Expected_utility_hypothesis), was placed on an axiomatic footing by [von Neumann and Morgenstern in 1944](https://en.wikipedia.org/wiki/Von_Neumann%E2%80%93Morgenstern_utility_theorem), and extended into modern equilibrium theory by [Samuelson](https://en.wikipedia.org/wiki/Paul_Samuelson) and [Debreu](https://en.wikipedia.org/wiki/G%C3%A9rard_Debreu). Think of a utility function as a mathematical representation of satisfaction or happiness. It assigns a real number to each possible portfolio, with higher numbers corresponding to more preferred portfolios. The investor's goal is to choose the portfolio that maximizes their utility (happiness), subject to constraints like their budget.

Sara's pitch is to apply classical consumer choice machinery to portfolio allocation, with the AI layer showing up only in how preferences are computed each trading period, e.g., seconds, minutes, days, etc. Let's look at the general problem that are trying to solve:

> __The general problem:__
>
> At time $t$, given $N$ assets with current prices $P_i \equiv S_i(t)$ for $i=1,\ldots,N$ and a total investment budget $B \equiv W_{\mathcal{P}}(t)$, choose the number of shares $n_1, \ldots, n_N$ that __maximizes the investor's utility__ subject to the __budget constraint__ that the total cost of the portfolio equals the budget:
>
> $$\operatorname*{maximize}_{n_1, \ldots, n_N} \quad U(n_1, \ldots, n_N) \qquad \text{subject to} \quad \sum_{i=1}^{N} n_i \cdot P_i = B$$
>
> The utility function $U: \mathbb{R}^N \to \mathbb{R}$ encodes the investor's preferences. Different allocation strategies, i.e., different values for $n_1, \ldots, n_N$, correspond to different values of the utility function. Higher utility corresponds to more preferred portfolios. The budget constraint ensures that the total cost of the portfolio does not exceed the available wealth.

In session 2 (and tonight) we'll explore one specific utility function, the Cobb-Douglas utility, that has a clean analytical solution and encodes preferences naturally through exponents. 

### Cobb-Douglas Utility
The **Cobb-Douglas utility function** is the workhorse of this course. It has a clean analytical solution, encodes preferences naturally through exponents, and separates assets into _preferred_ and _non-preferred_ categories.

> __Cobb-Douglas Utility__
>
> Given $N$ assets with preference weights $\gamma_i\in\mathbb{R}$, the Cobb-Douglas utility function is given by:
> $$U(n_1, \ldots, n_N) = \kappa(\boldsymbol{\gamma}) \prod_{i=1}^{N} n_i^{\gamma_i}$$
>
> where $\gamma_i \in (-1, 1)$ is the preference exponent for asset $i$, $n_i$ is the number of shares of asset $i$, and $\kappa = +1$ if all $\gamma_i \geq 0$, else $\kappa = -1$. The preference exponents do all the work:
> * $\gamma_i > 0$ **preferred**: higher $\gamma_i$ means stronger preference for asset $i$.
> * $\gamma_i \leq 0$ **non-preferred**: asset $i$ receives the minimum position $\epsilon$.
>
> The product-of-powers structure yields a clean result under budget constraint: the optimal _budget share_ of each preferred asset is proportional to its exponent $\gamma_i$ directly (not raised to a power). The budget-constrained Cobb-Douglas problem has a closed-form analytical solution. Partition assets into preferred $\mathcal{A}^+ = \{i : \gamma_i > 0\}$ and non-preferred $\mathcal{A}^- = \{i : \gamma_i \leq 0\}$ sets. The optimal number of shares is given by:
> $$n_i^{\star} = \underbrace{\left(\frac{\gamma_i}{\sum_{j \in \mathcal{A}^+} \gamma_j}\right)}_{\text{preference share}} \cdot \frac{B_{\text{adj}}}{P_i} \quad \text{for } i \in \mathcal{A}^+, \qquad n_i^{\star} = \epsilon \quad \text{for } i \in \mathcal{A}^-$$
>
> where $B_{\text{adj}} = B - \epsilon \sum_{k \in \mathcal{A}^-} P_k$ is the budget remaining after funding minimum positions. The allocation responds immediately to changes in $\gamma_i$; no matrix inversion, no quadratic program, no numerical solver.

The Cobb-Douglas allocator needs preference weights $\gamma_i$ as input. Where do they come from? From a combination of **fundamental signals** (the Single Index Model) and a **sentiment signal** $\lambda_t$. There are many ways to compute the preference weights, and the sentiment signal. However, let's look at one specific formulation that captures the intuition.

> __Preference Weight Computation:__
>
> We propose a preference model of the form:
>
> $$\gamma_i = \sigma\!\left(\boldsymbol{\theta}_{i}^{\top}\mathbf{x}_{t}\right)\quad\text{for } i=1,\ldots,N$$
>
> where $\sigma$ is a link function, $\boldsymbol{\theta}_i$ are per-asset parameters (estimated from data or set by hand), and $\mathbf{x}_{t}$ are time-dependent features (fundamental signals, technical indicators, macroeconomic variables).
>
> For this course, let the **SIM parameters** for asset $i$ be $(\alpha_i, \beta_i, \sigma_{\varepsilon,i})$ from Session 1, so $\mathbb{E}[g_i] = \alpha_i + \beta_i\,\mathbb{E}[g_{\mathrm{mkt}}]$. We take $\sigma = \tanh$ modulated by a sentiment scalar $\lambda_t$ that penalizes high-beta names in bearish regimes ($\lambda_t > 0$) and amplifies them in bullish ones ($\lambda_t < 0$). Dividing each SIM term by $\beta_i^{\lambda_t}$ gives:
>
> $$\boxed{\gamma_i = \tanh\!\left(\frac{\alpha_i}{\beta_i^{\lambda_t}} + \beta_i^{1-\lambda_t} \cdot \mathbb{E}[g_{\mathrm{mkt}}]\right)\quad\blacksquare}$$
>
> so $\boldsymbol{\theta}_i = (\alpha_i, \beta_i)$ and $\mathbf{x}_t = (1,\ \mathbb{E}[g_{\mathrm{mkt}}])$, with $\beta_i^{\lambda_t}$ as a *regime lens* on those coefficients rather than extra parameters. Adding a news feature $\text{news}_i$ (a per-period LLM-aggregated sentiment scalar per ticker) with loading $\nu_i$ extends the design as follows:
>
> $$\gamma_i = \tanh\!\left(\frac{\alpha_i}{\beta_i^{\lambda_t}} + \beta_i^{1-\lambda_t}\,\mathbb{E}[g_{\mathrm{mkt}}] + \nu_i\,\text{news}_i\right)$$
>
> The news term sits outside the regime-lens wrapper: $\lambda_t$ interprets market exposure, while $\nu_i\,\text{news}_i$ adds idiosyncratic information that did not enter the SIM regression.

How do we compute the sentiment signal $\lambda_t$? Many approaches work. We propose a simple exponential moving average (EMA) crossover signal that captures short-term momentum as a proxy for market sentiment. Let's take a look at how that works.

> __The Lambda Signal:__
>
> The exponential moving average (EMA) of a market index, e.g., SPY, QQQ, IWM, etc price series $S_t$ (USD/share, e.g., SPY) with window-size $L$ (days) is given by:
>
> $$\bar{S}_{t} = \omega\, S_{t} + (1 - \omega)\,\bar{S}_{t-1}, \qquad \omega = \frac{2}{L + 1}$$
>
> For this strategy we compute two EMAs: a **short-term** ($L_{\mathrm{short}} = 21$ days) that tracks recent momentum, and a **long-term** ($L_{\mathrm{long}} = 63$ days) that tracks the underlying trend. The crossover between short and long EMAs produces the sentiment signal:
>
> $$\lambda_t = -G \cdot \left(\frac{\bar{S}^{\text{short}}_t}{\bar{S}^{\text{long}}_t} - 1\right)$$
>
> where $G > 0$ is a gain constant that controls sensitivity (hyperparameter). The interpretation of $\lambda_t$ is given by:
> * $\lambda_t > 0$: **bearish** (short EMA below long EMA). The engine becomes risk-averse.
> * $\lambda_t < 0$: **bullish** (short EMA above long EMA). The engine takes on more risk.
> * $\lambda_t \approx 0$: **neutral** (no strong signal).
>
> The EMA crossover is one choice of sentiment signal. Other sources, such as natural language processing (NLP) derived scores from news articles, earnings calls, or regulatory filings, can also drive $\lambda_t$.

## The Problem Tonight: Three Inputs Feed $\gamma_i$, and All Three Are Incomplete

In Session 2, the engine computes preference weights $\gamma_i$ from SIM parameters $(\alpha_i, \beta_i, \sigma_{\varepsilon,i})$ estimated once from a historical training window, runs on a universe of tickers picked once from the Session 1 configuration, and consumes a news-aware $\gamma_i$ formula whose news term has never been wired to a real signal source. Three of the same engine's inputs are open at the start of Session 3.

> __Three gaps in the Session 2 engine:__
>
> 1. **Parameter drift.** The SIM triple $(\alpha_i, \beta_i, \sigma_{\varepsilon,i})$ was calibrated from a training window and never updated. If the market enters a new regime, the frozen parameters send stale signals to the preference weight formula. Alpha estimates from a bull market are misleading in a bear market; beta estimates from low-volatility periods understate risk during a crisis.
> 2. **Hand-picked universe.** The set of tickers came from a configuration file. The configuration was built once with batch SIM estimates and a portfolio interview; nothing in the engine ever asks whether a different subset of the same candidate pool would have produced higher Cobb-Douglas utility. The selection is fully detached from the realized prices the engine sees during the backtest.
> 3. **Missing news channel.** Session 2 declared the news-aware preference weight $\gamma_i = \tanh(\alpha_i / \beta_i^{\lambda_t} + \beta_i^{1-\lambda_t}\,\mathbb{E}[g_{\mathrm{mkt}}] + \nu_i\,\text{news}_i)$ but never said where $\text{news}_i$ comes from at production time or how to estimate $\nu_i$ from data. The formula describes a feature the engine cannot read.

Session 3 closes all three gaps with classical reinforcement-learning machinery and an LLM-driven news pipeline. We use a multi-armed bandit over $2^{K}-1$ ticker subsets to learn the universe, an LLM-driven news pipeline that scores headlines via Claude and recovers the loading $\nu_i$ via OLS alongside $(\alpha_i, \beta_i)$, and exponentially weighted least squares to keep the SIM parameters $(\alpha_i, \beta_i, \sigma_{\varepsilon,i})$ updated online. The CES elasticity heuristic from Session 2 is also a hand-tuned mapping rather than a learned one; since the deterministic adaptive elasticity is what the production engine in Session 4 still consumes, we treat the elasticity-learning extension in the optional materials.

## Reinforcement Learning (RL)
What is reinforcement learning? The short answer is that it's a family of methods for learning how to make decisions in an environment to maximize cumulative reward. Just like us, the agent learns by doing. 


An agent in state $s \in \mathcal{S}$ takes an action $a \in \mathcal{A}$ and receives a reward $r$ from the environment, then transitions to a new state and repeats. Reinforcement learning is the family of methods that learns, from this loop, how to choose actions that maximize cumulative reward over time. The canonical text covering all three approaches below is [Sutton and Barto, *Reinforcement Learning: An Introduction* (2nd ed., 2018)](http://incompleteideas.net/book/the-book-2nd.html). Several distinct strategies fall under this umbrella:

* __Multiplicative weights__ adjusts the probability of selecting an action based on past performance, in a principled way that guarantees the algorithm performs nearly as well as the best fixed action in hindsight, even in changing environments; that is, it minimizes regret. See [Arora, Hazan, and Kale (2012), *The Multiplicative Weights Update Method: A Meta-Algorithm and Applications*](https://theoryofcomputing.org/articles/v008a006/) for the unified survey.
* __Bandit algorithms__ operate in stateless environments. On each round, they explore different actions to estimate their rewards and adapt their action-selection strategy based on the outcomes. The modern reference is [Slivkins, *Introduction to Multi-Armed Bandits* (2019)](https://arxiv.org/abs/1904.07272); the encyclopedic treatment is [Lattimore and Szepesvári, *Bandit Algorithms* (2020)](https://tor-lattimore.com/downloads/book/book.pdf).
* __Q-learning__ is a value-based method that estimates the long-term value (utility, satisfaction, happiness, etc) of each state-action pair, enabling the agent to learn optimal behavior in environments with temporal and sequential dynamics. The original convergence proof is [Watkins and Dayan, *Q-learning*, *Machine Learning* 8 (1992)](https://link.springer.com/article/10.1007/BF00992698); the deep-network extension that brought the method into mainstream practice is [Mnih et al., *Human-level control through deep reinforcement learning*, *Nature* 518 (2015)](https://www.nature.com/articles/nature14236).

All three must balance the same fundamental challenge: explore new actions to gather information, or exploit known actions to maximize reward.

__RL has impacted many domains__. [Mnih et al. (2015)](https://www.nature.com/articles/nature14236) trained Atari agents from pixels at human level and opened the deep-RL line that produced DeepMind's [AlphaGo Zero](https://www.nature.com/articles/nature24270) and [AlphaStar](https://www.nature.com/articles/s41586-019-1724-z); [DeepMind's data-center cooling system](https://deepmind.google/discover/blog/deepmind-ai-reduces-google-data-centre-cooling-bill-by-40/) cut cooling energy by roughly 30% on Google sites; and [Wei et al. (2018)](https://dl.acm.org/doi/10.1145/3219819.3220096) scheduled traffic-light phases in real time to reduce congestion. All of these predate the [transformer revolution](https://arxiv.org/abs/1706.03762) of 2017, which shifted the field toward large language models and reasoning agents; the tide may now be turning back toward RL as the next frontier.

__Is RL Back in Style?__ [DeepSeek-R1's training pipeline](https://arxiv.org/abs/2501.12948) relies primarily on reinforcement learning rather than human-labeled examples to elicit chain-of-thought behavior (the R1-Zero variant uses pure RL from a base model with rule-based rewards). In April 2026 [David Silver raised $1.1B for Ineffable Intelligence](https://techcrunch.com/2026/04/27/deepminds-david-silver-just-raised-1-1b-to-build-an-ai-that-learns-without-human-data/) to pursue [agents that learn without human-generated data](https://www.ineffable.ai). Alternative visions are also gaining traction: [Yann LeCun raised $1B to found AMI (Advanced Machine Intelligence)](https://www.linkedin.com/company/amilabs-xyz) to pursue [Joint-Embedding Predictive Architecture (JEPA)](https://arxiv.org/abs/2601.00844) systems. [What is a JEPA?](https://www.yout-ube.com/watch?v=kYkIdXwW2AE)

## Multi-Armed Bandits

The Session 1 universe was hand-picked once from the portfolio interview and never revisited. A bandit treats universe selection as a learning problem: choose a subset of the $K$ candidate tickers at each rebalance, observe the realized Cobb-Douglas utility, and converge on the subset that maximizes expected utility. The same machinery generalizes to other discrete-action choices the engine faces (CES elasticity is a small-scale example we walk later), so we lay down the framework in full generality first.

> __The Multi-Armed Bandit Framework:__
>
> A bandit faces a finite arm (choice) set $\mathcal{A} = \{1, 2, \ldots, K\}$ of cardinality $K$. At each round $t = 1, \ldots, T$ the learner selects an arm $a_t \in \mathcal{A}$ and observes a stochastic reward $r_t$ drawn from a fixed but unknown distribution that depends on $a_t$. Let $\mu_k = \mathbb{E}[r_t \mid a_t = k]$ denote the mean reward of arm $k$ and $\mu^{\star} = \max_{k} \mu_k$ the mean of the optimal arm. The cumulative gap between $\mu^{\star}$ and the means of the arms actually pulled is the learner's *regret*: minimizing regret is equivalent to identifying the best arm fast and exploiting it.
>
> The **epsilon-greedy** policy with a decaying exploration rate $\varepsilon_t \to 0$ explores a random arm with probability $\varepsilon_t$ and otherwise pulls the arm with the highest estimated mean. The pseudocode below uses the standard $t^{-1/3}(K \ln t)^{1/3}$ schedule that achieves provably sublinear regret (finite-time analysis: Auer, Cesa-Bianchi, and Fischer [2002]; modern treatment: [Slivkins, *Introduction to Multi-Armed Bandits*](https://arxiv.org/abs/1904.07272)). The learner converges, but never stops exploring entirely.

This is an *online* algorithm: reward draws arrive one at a time, and the arm-mean estimates update after every pull rather than at the end of a batch. The mechanics are compact enough to state as pseudocode directly.

### Algorithm: Epsilon-Greedy Bandit (Online)

__Initialize__: Given an arm set $\mathcal{A} = \{1, 2, \ldots, K\}$, a horizon $T$, and a reward oracle that returns a stochastic reward $r_t$ when an arm is pulled, set the mean-reward estimates $\hat{\mu}_a \gets 0$ and pull counts $n_a \gets 0$ for each $a \in \mathcal{A}$.

For $t = 1, \ldots, T$ __do__:

1. Compute the exploration rate: $\varepsilon_t \gets t^{-1/3} \cdot (K \ln t)^{1/3}$.
2. Draw $p \sim \mathrm{Uniform}[0,1]$ and select an arm:
    - If $p \leq \varepsilon_t$, *explore*: sample $a_t$ uniformly from $\mathcal{A}$.
    - Otherwise, *exploit*: set $a_t \gets \arg\max_{a \in \mathcal{A}} \hat{\mu}_a$ (break ties arbitrarily).
3. Pull arm $a_t$ and observe the stochastic reward $r_t$ from the environment.
4. Update the pull count and running mean for arm $a_t$:
    $$n_{a_t} \gets n_{a_t} + 1, \qquad \hat{\mu}_{a_t} \gets \hat{\mu}_{a_t} + \frac{1}{n_{a_t}}\left(r_t - \hat{\mu}_{a_t}\right)$$

__Output__: Return the mean-reward estimates $\{\hat{\mu}_a\}_{a \in \mathcal{A}}$, pull counts $\{n_a\}_{a \in \mathcal{A}}$, and the action-reward history $\{(a_t, r_t)\}_{t=1}^{T}$.

The sample-average update in step 4 gives every observation of arm $a$ equal weight; replacing $1/n_{a_t}$ with a constant learning rate $\alpha \in (0,1)$ makes the estimate track non-stationary rewards at the cost of residual variance.

### Apply the Bandit to Ticker Selection
Let's use a bandit to solve the universe-selection problem: which subset of the candidate tickers should the engine trade? Session 1 picked the basket once from the portfolio interview and never revisited it. We now treat universe selection as a learning problem in which the bandit selects a subset, observes the Cobb-Douglas utility realized on that subset over the next rebalance window, and converges on the subset that maximizes expected utility.

> __Ticker Selection Bandit:__
>
> * **Arms:** the $2^{K} - 1$ non-empty subsets of a candidate universe of $K$ tickers. Each arm is a basket; selecting an arm tells the engine which tickers are eligible to receive non-zero share counts, and the Cobb-Douglas allocator solves for the optimal positions inside the chosen basket. The combinatorial cardinality grows fast: $K = 8$ gives $255$ arms, $K = 12$ gives $4{,}095$, and $K = 22$ gives $\approx 4.2$ million, which is what motivates the optional DQN extension that we explore later.
> * **Reward:** the Cobb-Douglas utility $U_{\mathrm{CD}}(\mathbf{n}^{\star}(\mathcal{S}))$ realized on the selected subset $\mathcal{S} \subseteq \{1, \ldots, K\}$, where $\mathbf{n}^{\star}(\mathcal{S})$ is the optimal share-count vector returned by the Cobb-Douglas allocator at current prices and preference weights restricted to $\mathcal{S}$.
> * **Comparison baseline:** the hand-picked Session 1 universe held fixed across the entire backtest. This is the no-learning baseline the bandit-selected subset has to beat on paired-path wealth and Sharpe.

So how could we solve this problem? The combinatorial action space is large, but the reward is observed immediately after each selection, so we can apply the epsilon-greedy bandit directly to the set of subsets. 

### Algorithm: Combinatorial Ticker Bandit (Online)

__Initialize__: Given a candidate universe of $K$ tickers, the action set $\mathcal{A} = \{\mathcal{S} \subseteq \{1, \ldots, K\} : \mathcal{S} \neq \emptyset\}$ of cardinality $|\mathcal{A}| = 2^{K} - 1$, a rebalance schedule $\{b_t\}_{t=1}^{T}$ with $b_t \in \{0, 1\}$, and a Cobb-Douglas utility oracle $U_{\mathrm{CD}}(\mathbf{n}^{\star}(\mathcal{S}))$ that returns the realized utility when the allocator runs on subset $\mathcal{S}$, set a round counter $\tau \gets 0$, and for each $\mathcal{S} \in \mathcal{A}$ initialize $\hat{\mu}_{\mathcal{S}} \gets 0$ and $n_{\mathcal{S}} \gets 0$.

For $t = 1, \ldots, T$ with $b_t = 1$ __do__:

1. Increment the round counter and compute the exploration rate: $\tau \gets \tau + 1$, $\varepsilon \gets \tau^{-1/3} \cdot (|\mathcal{A}| \ln \tau)^{1/3}$.
2. Draw $p \sim \mathrm{Uniform}[0,1]$ and select an arm:
    - If $p \leq \varepsilon$, *explore*: sample $\mathcal{S}_t$ uniformly from $\mathcal{A}$.
    - Otherwise, *exploit*: set $\mathcal{S}_t \gets \arg\max_{\mathcal{S} \in \mathcal{A}} \hat{\mu}_{\mathcal{S}}$.
3. Solve the Cobb-Douglas allocation on subset $\mathcal{S}_t$ and observe the reward $r_t \gets U_{\mathrm{CD}}(\mathbf{n}^{\star}(\mathcal{S}_t))$.
4. Update the pulled arm:
    $$n_{\mathcal{S}_t} \gets n_{\mathcal{S}_t} + 1, \qquad \hat{\mu}_{\mathcal{S}_t} \gets \hat{\mu}_{\mathcal{S}_t} + \frac{1}{n_{\mathcal{S}_t}}\left(r_t - \hat{\mu}_{\mathcal{S}_t}\right)$$

__Output__: Return the learned subset $\mathcal{S}^{\star} = \arg\max_{\mathcal{S} \in \mathcal{A}} \hat{\mu}_{\mathcal{S}}$, the arm means $\{\hat{\mu}_{\mathcal{S}}\}$, pull counts $\{n_{\mathcal{S}}\}$, and the chosen-arm history $\{(\mathcal{S}_t, r_t)\}$.

At large $K$ this approach runs out of room: $K = 22$ already pushes us past four million arms, more than any realistic exploration budget can visit even once. The DQN extension in the optional materials replaces the table with a function approximator that generalizes across baskets, which is what makes large universes tractable.

We train an epsilon-greedy bandit over $2^K - 1$ ticker subsets with Cobb-Douglas utility as the reward, deploy the chosen subset across the Monte Carlo ensemble, and compare paired-path wealth and Sharpe to the hand-picked Session 1 baseline.

### Applying the Bandit to CES Elasticity
The same bandit machinery applies to a second, smaller-scale problem: picking the CES elasticity $\eta$ that the allocator runs at, from a discrete grid with one bandit per sentiment regime. This is a side track relative to the production engine (which uses the deterministic Session 2 heuristic for $\eta$), but it's a clean illustration of the bandit framework on a small arm set, $K \approx 5$, where the mechanics are easier to inspect than in the combinatorial ticker case above.

> __CES Utility and Elasticity Bandit:__
>
> The Constant Elasticity of Substitution (CES) utility generalizes Cobb-Douglas with an elasticity parameter $\eta > 0$ that sets how willingly the investor substitutes between assets:
>
> $$U_{\text{CES}}(\mathbf{n}) = \left(\sum_{i \in \mathcal{A}^+} \gamma_i \cdot n_i^{\rho}\right)^{1/\rho}, \qquad \rho = \frac{\eta - 1}{\eta}$$
>
> Unlike Cobb-Douglas, $\gamma_i$ enters as a **coefficient** on $n_i^{\rho}$ rather than an exponent; only positive weights $\gamma_i \in (0, 1)$ contribute (non-preferred names still get the $\varepsilon$-floor), and CES has no closed form so we solve numerically. Three limits frame the family along the **bang-for-the-buck** ratio $\gamma_i / P_i$:
> * $\eta \to 1$: collapses to Cobb-Douglas; dollar weights $w_i^{\star} = \gamma_i / \sum_j \gamma_j$ track preferences and ignore prices.
> * $\eta \to \infty$: linear utility; allocator concentrates on the single highest $\gamma_i / P_i$.
> * $\eta \to 0^{+}$: Leontief (perfect complements); equal share counts independent of $\gamma_i$ or $P_i$.
>
> Proofs in the companion [CES limits derivation notebook](deeper/eCornell-AI-Finance-S2-Derivation-CES-Limits-May-2026.ipynb). A fixed $\eta$ ignores the market; linking it to the same sentiment signal $\lambda_t$ that drives preferences gives an expression of the form:
>
> $$\boxed{\eta(\lambda_t) = \eta_{\min} + \frac{\eta_{\max} - \eta_{\min}}{1 + |\lambda_t|}\quad\blacksquare}$$
>
> The $|\lambda_t|$ in the denominator makes the response symmetric: at $\lambda_t \approx 0$ the allocator concentrates ($\eta \to \eta_{\max}$); at extreme conviction either way it diversifies ($\eta \to \eta_{\min}$), hedging against the chance the signal is noise. **Defaults** $\eta_{\min} = 0.5$ and $\eta_{\max} = 5.0$ keep the allocator preference-responsive at the low end and concentration-capped at the high end. In short: $\gamma_i$ picks *which* assets, $\eta(\lambda_t)$ sets *how aggressively*.
>
> But could a bandit learn a better mapping from sentiment to elasticity? 
>
> __Bandit setup:__
>
> How would we set up the bandit to learn an improved elasticity heuristic? Here's one way:
> * **Arms:** a finite ordered grid $\mathcal{H} = \{\eta_1 < \cdots < \eta_K\}$ on $[\eta_{\min}, \eta_{\max}]$, near-Leontief at the low end and near-linear at the high end. Cardinality $K$ and spacing are design choices; the optional example fixes a specific grid, but you can experiment with different ones.
> * **Reward:** CES utility $U_{\mathrm{CES}}(\mathbf{n}^{\star}(\eta))$ at the chosen $\eta$, where $\mathbf{n}^{\star}(\eta)$ is the optimal share-count vector from the CES allocator at current prices and preferences.
> * **Per-regime learning:** Partition $\lambda_t$ via threshold $\theta$ into bearish ($\lambda_t > \theta$), neutral ($|\lambda_t| \leq \theta$), and bullish ($\lambda_t < -\theta$); each bin runs its own bandit and learns its own best $\eta \in \mathcal{H}$, since a single global bandit would converge on the regime-averaged $\eta$ and miss the conditional structure.
> * **Comparison baseline:** The Session 2 heuristic $\eta(\lambda_t) = \eta_{\min} + (\eta_{\max} - \eta_{\min})/(1 + |\lambda_t|)$.
>
> The per-regime partition lets the bandit discover asymmetric structure. A bearish regime might call for aggressive diversification (small $\eta$) while a bullish regime tolerates concentration (large $\eta$). The heuristic treats all extreme sentiment identically because it depends on $|\lambda|$; the bandit can break this symmetry.

### Algorithm: Per-Regime Elasticity Bandit (Online)

__Initialize__: Given an elasticity grid $\mathcal{H} = \{\eta_1, \ldots, \eta_K\}$, a sentiment threshold $\theta > 0$, a regime set $\mathcal{R} = \{\mathrm{bear}, \mathrm{neutral}, \mathrm{bull}\}$, a sentiment series $\{\lambda_t\}_{t=1}^{T}$, a rebalance schedule $\{b_t\}_{t=1}^{T}$ with $b_t \in \{0, 1\}$, and a CES utility oracle $U_{\mathrm{CES}}(\mathbf{n}^{\star}(\eta))$ that returns the realized utility when the allocator runs at elasticity $\eta$, initialize a bandit per regime: for each $\rho \in \mathcal{R}$ set a regime round counter $\tau_\rho \gets 0$, and for each $(\rho, \eta) \in \mathcal{R} \times \mathcal{H}$ set $\hat{\mu}_{\rho,\eta} \gets 0$ and $n_{\rho,\eta} \gets 0$.

For $t = 1, \ldots, T$ with $b_t = 1$ __do__:

1. Bin the current sentiment $\lambda_t$ into a regime:
    - If $\lambda_t > \theta$, set $\rho_t \gets \mathrm{bear}$.
    - Else if $\lambda_t < -\theta$, set $\rho_t \gets \mathrm{bull}$.
    - Otherwise, set $\rho_t \gets \mathrm{neutral}$.
2. Advance the regime round counter and compute the regime-local exploration rate: $\tau_{\rho_t} \gets \tau_{\rho_t} + 1$, $\varepsilon \gets \tau_{\rho_t}^{-1/3} \cdot (K \ln \tau_{\rho_t})^{1/3}$.
3. Draw $p \sim \mathrm{Uniform}[0,1]$ and select an arm from bandit $\rho_t$:
    - If $p \leq \varepsilon$, *explore*: sample $\eta_t$ uniformly from $\mathcal{H}$.
    - Otherwise, *exploit*: set $\eta_t \gets \arg\max_{\eta \in \mathcal{H}} \hat{\mu}_{\rho_t, \eta}$.
4. Solve the CES allocation at elasticity $\eta_t$ and observe the reward $r_t \gets U_{\mathrm{CES}}(\mathbf{n}^{\star}(\eta_t))$.
5. Update the pulled arm for the active regime:
    $$n_{\rho_t, \eta_t} \gets n_{\rho_t, \eta_t} + 1, \qquad \hat{\mu}_{\rho_t, \eta_t} \gets \hat{\mu}_{\rho_t, \eta_t} + \frac{1}{n_{\rho_t, \eta_t}}\left(r_t - \hat{\mu}_{\rho_t, \eta_t}\right)$$

__Output__: Return the learned elasticity map $\rho \mapsto \arg\max_{\eta \in \mathcal{H}} \hat{\mu}_{\rho, \eta}$ for $\rho \in \mathcal{R}$, together with the per-regime arm means $\{\hat{\mu}_{\rho,\eta}\}$, pull counts $\{n_{\rho,\eta}\}$, and the chosen-arm history $\{(\rho_t, \eta_t, r_t)\}$.

## Review: Q-Learning Theory
Q-learning estimates the action-value function $Q(s, a)$ by conducting repeated experiments $t=1,2,\ldots$ in the world $\mathcal{W}$. 
In each experiment, an agent in state $s\in\mathcal{S}$ takes action $a\in\mathcal{A}$, receives a reward $r$, and (potentially) transitions to a new state $s^{\prime}$. After each experiment $t$, the agent updates its estimate of $Q(s, a)$ using the update rule:
$$
\begin{equation*}
Q_{t+1}(s,a)\leftarrow{\underbrace{Q_{t}(s,a)}_{\text{old value}}}+\alpha_{t}\cdot\underbrace{\left(r+\gamma\cdot\max_{a^{\prime}\in\mathcal{A}}Q_{t}(s^{\prime},a^{\prime}) - Q_{t}(s,a)\right)}_{\text{TD error}}\quad{t = 1,2,3,\ldots}
\end{equation*}
$$
where $0<\alpha_{t} <{1}$ is the learning rate parameter at time $t$, and $0<\gamma<{1}$ is the discount factor. 
We estimate the policy function $\pi:\mathcal{S}\rightarrow\mathcal{A}$ by selecting the action $a$ that maximizes $Q(s,a)$ at each state $s$:
$$
\begin{equation*}
\pi(s) = \arg\max_{a\in\mathcal{A}}Q(s,a)
\end{equation*}
$$


### Algorithm
__Initialize__ $Q(s,a)$ arbitrarily for all $s\in\mathcal{S}$ and $a\in\mathcal{A}$. Set the hyperparameters: initial learning rate $\alpha_{0}\in(0,1)$ with decay factor $\beta\in(0,1]$, discount factor $\gamma$, exploration schedule, maximum iteration count $\texttt{maxiter}$, convergence tolerance $\delta$, and convergence window $W$. Set $\texttt{converged}\gets\texttt{false}$.

- For each $s\in\mathcal{S}$:
    - Set the trial counter $t\gets 1$.
    - While $\texttt{converged}$ is $\texttt{false}$:
        1. Roll a random number $p\in[0,1]$. Compute $\varepsilon_{t}=\min\{1,\,t^{-1/3}\cdot\bigl(K\cdot\log(t+1)\bigr)^{1/3}\}$, where $K=|\mathcal{A}|$ is the number of actions. The clip to $1$ keeps $\varepsilon_{t}$ a valid probability for small $t$, and $\log(t+1)$ avoids the degenerate $\varepsilon_{1}=0$.
        2. If $p\leq\varepsilon_{t}$, choose a random (uniform) action $a_{t}\in\mathcal{A}$. Otherwise, choose the greedy action $a_{t}=\arg\max_{a\in\mathcal{A}}Q_{t}(s,a)$.
        3. Take action $a_{t}$, observe the reward $r$ from the __world__, and transition to the next state $s^{\prime}$.
        4. Update the state-action-value function: $Q_{t+1}(s,a)\leftarrow Q_{t}(s,a)+\alpha_{t}\cdot\bigl(r+\gamma\cdot\max_{a^{\prime}\in\mathcal{A}}Q_{t}(s^{\prime},a^{\prime})-Q_{t}(s,a)\bigr)$.
        5. Update the state $s\leftarrow s^{\prime}$, decay the learning rate $\alpha_{t+1}\leftarrow\beta\,\alpha_{t}$, and increment $t\leftarrow t+1$.
        6. Check stopping conditions:
            - If the largest update over the last $W$ iterations, $\Delta_{t}=\max_{k\in\{t-W,\ldots,t\}}|Q_{k+1}(s,a)-Q_{k}(s,a)|$, satisfies $\Delta_{t}\leq\delta$, set $\texttt{converged}\gets\texttt{true}$.
            - Else if $t\geq\texttt{maxiter}$, set $\texttt{converged}\gets\texttt{true}$ and notify the caller that the iteration limit was reached without convergence.
            - Otherwise, continue to the next iteration.
    - End While
- End For

__Output__: Return the estimated state-action-value function $Q(s,a)$ and the derived policy $\pi(s) = \arg\max_{a\in\mathcal{A}}Q(s,a)$ for all $s\in\mathcal{S}$.

## Advanced: Deep Q-Learning (DQN)
Deep Q-learning replaces the Q-table with a neural network that approximates the Q-value function. The network maps a state to a vector of Q-values, one per action, which sidesteps the [curse of dimensionality](https://en.wikipedia.org/wiki/Curse_of_dimensionality) that makes tabular Q-learning infeasible for image-like or continuous inputs.

For our universe-selection problem, the bandit's table strategy collapses fast. With $N$ tickers and a portfolio of any subset, the bandit faces $2^{N} - 1$ arms; every basket is an opaque, unrelated arm, and the table size grows exponentially in $N$. With $N = 22$ we are looking at roughly $4.2$ million arms; with $N = 50$, about $10^{15}$. No realistic exploration budget visits each arm even once. DQN replaces the table with a function approximator $Q_{\boldsymbol{\theta}}(s) \in \mathbb{R}^{|\mathcal{A}|}$ whose shared parameters generalize across states: what we learn from _AAPL + MSFT_ transfers to _AAPL + GOOG_ because both share most input features. Framed sequentially, where we pick one ticker per step and have the network score the remaining candidates, the action space collapses from $2^{N}$ subsets to $N$ per-ticker scores.

The next subsection writes down the four moving parts (Q-network, target network, replay buffer, $\varepsilon$-greedy policy) and the algorithm that ties them together.

### Theory of DQN
A deep Q-learning agent learns a policy $\pi$ that maximizes the expected cumulative reward $R_t$ over time. Suppose the agent is tasked with making decisions over $T\rightarrow\infty$ steps.

For each episode, we sample for $t = 1,2,\ldots,T$: 

1. __Interaction with the environment__: At each time step $t$, the agent observes the current state $s_t$, selects an action $a_t$ (typically using an $\varepsilon$-greedy policy based on the _Q-network_), and receives a reward $r_t$ and the next state $s_{t+1}$ from the environment.
2. __Experience replay__: Each transition tuple $(s_t, a_t, r_t, s_{t+1})$ is stored in a **replay buffer** (a finite-sized memory that we'll use for training). Instead of training on consecutive samples, the agent **samples random mini-batches** from this buffer. 
3. __Main Q-Network (function approximator)__: The core of DQN is a deep neural network $Q_{\boldsymbol{\theta}}(s)\in\mathbb{R}^{|\mathcal{A}|}$ with (trainable) parameters $\boldsymbol{\theta}$, which learns to approximate the optimal action-value function. The network takes a state as input and outputs a vector of Q-values, one per action; we write $[Q_{\boldsymbol{\theta}}(s)]_{a}$ for the entry corresponding to action $a$.
4. __Target Q-Network__: To stabilize training, DQN uses a **target network** $Q^{\prime}_{\boldsymbol{\theta}^{-}}(s)\in\mathbb{R}^{|\mathcal{A}|}$, which is a delayed copy of the main Q-network. The target network's parameters $\boldsymbol{\theta}^-$ are updated periodically (e.g., every $C$ steps) by copying the weights from the main Q-network.

These four parts come together in a single loss surface. For each transition $(s_{i}, a_{i}, r_{i}, s_{i+1}, d_{i})$ in the replay buffer, the bootstrap target uses the target network to score the best next action, and the main network is regressed onto that target on the action that was actually taken:

> __DQN Bellman Loss:__
>
> The bootstrap target for transition $i$ is given by:
>
> $$y_{i} = r_{i} + \gamma\,(1 - d_{i})\,\max_{a^{\prime} \in \mathcal{A}}\,[Q^{\prime}_{\boldsymbol{\theta}^{-}}(s_{i+1})]_{a^{\prime}}$$
>
> where $\gamma \in [0, 1)$ is the discount factor and $(1 - d_{i})$ zeros the bootstrap on terminal transitions. The Q-network is trained by stochastic gradient descent on the mean squared Bellman residual:
>
> $$\boxed{\;\mathcal{L}(\boldsymbol{\theta}) \;=\; \frac{1}{B}\sum_{i = 1}^{B}\Bigl(y_{i} - [Q_{\boldsymbol{\theta}}(s_{i})]_{a_{i}}\Bigr)^{2}\quad\blacksquare\;}$$
>
> The two halves of the residual are deliberately asymmetric: the prediction $[Q_{\boldsymbol{\theta}}(s_{i})]_{a_{i}}$ uses the *action that was actually taken* in transition $i$, while the target $y_{i}$ uses the *best action available next* under the frozen target network. The target network and the replay buffer are the two stabilizers; the target network keeps the bootstrap target steady between syncs, and the replay buffer decorrelates consecutive transitions.

#### DQN Algorithm

__Initialize__ the parameters of the main Q-network $Q_{\boldsymbol{\theta}}(s)$ to random values, and copy them to the target Q-network so that $\boldsymbol{\theta}^{-}\leftarrow\boldsymbol{\theta}$. Initialize an empty replay buffer $\mathcal{B}$ with maximum size $M$. Set the hyperparameters: the learning rate $\alpha>0$, the discount factor $\gamma\in(0,1)$, the warm-up threshold $N_{\text{warm}}$ (minimum buffer size before training begins), the mini-batch size $B$, and the target-network sync interval $C$.
- For each episode, initialize the state to $s_{0}$ and:
   - For each time step $t=1,\ldots,T$:
        1. Compute the exploration rate $\varepsilon_{t} \gets \min\{1,\, t^{-1/3}(|\mathcal{A}|\log(t+1))^{1/3}\}$ and roll a random number $p\in[0,1]$. If $p\leq\varepsilon_{t}$, choose a random (uniform) action $a_{t}\in\mathcal{A}$; otherwise, choose a greedy action $a_{t} = \arg\max_{a\in\mathcal{A}}\,[Q_{\boldsymbol{\theta}}(s_{t})]_{a}$.
        2. Execute action $a_{t}$, observe the reward $r_{t}$ and next state $s_{t+1}$ from the _world_, and observe a done flag $d_{t}\in\{0,1\}$ that is $d_{t} = 1$ if $s_{t+1}$ is terminal and $0$ otherwise.
        3. Store the transition $e_{t}=(s_{t}, a_{t}, r_{t}, s_{t+1}, d_{t})$ in the replay buffer: $\mathcal{B}\leftarrow\mathcal{B}\cup\{e_{t}\}$, evicting the oldest transition if $|\mathcal{B}|>M$.
        4. __Training step (only if $|\mathcal{B}|\geq N_{\text{warm}}$):__
            1. Sample a mini-batch of $B$ transitions $\{(s_{i}, a_{i}, r_{i}, s_{i+1}, d_{i})\}_{i=1}^{B}$ uniformly at random from $\mathcal{B}$.
            2. Compute the _target Q-value_ for each transition using the _target Q-network_: $y_{i} = r_{i} + \gamma\,(1 - d_{i})\,\max_{a^{\prime}\in\mathcal{A}}\,[Q^{\prime}_{\boldsymbol{\theta}^{-}}(s_{i+1})]_{a^{\prime}}$ for $i=1,2,\ldots,B$. The factor $(1 - d_{i})$ zeroes out the bootstrap term on terminal transitions.
            3. Compute the _mean squared loss_ over the $B$ experiences in the mini-batch using the action that was actually taken: $L(\boldsymbol{\theta}) = \frac{1}{B}\sum_{i=1}^{B}\left(y_{i} - [Q_{\boldsymbol{\theta}}(s_{i})]_{a_{i}}\right)^{2}$.
            4. Perform a _single_ gradient descent step to minimize the loss function $L(\boldsymbol{\theta})$ with respect to the parameters $\boldsymbol{\theta}$ of the main Q-network: $\boldsymbol{\theta} \leftarrow \boldsymbol{\theta} - \alpha\,\nabla_{\boldsymbol{\theta}}L(\boldsymbol{\theta})$.
                - _Why only a single step_? Each mini-batch is just a small sample of the environment's dynamics. The goal of DQN is _online learning_: the network parameters are continuously updated as new experiences come in. Forcing training to converge on each mini-batch risks _overfitting to that mini-batch_.
        5. Update the state $s_{t} \leftarrow s_{t+1}$.
        6. Every $C$ steps, update the target Q-network parameters: $\boldsymbol{\theta}^{-} \leftarrow \boldsymbol{\theta}$.
    - End For
- End For

The deployment-time policy is greedy: at each state $s$, take $a^{\star}(s) = \arg\max_{a \in \mathcal{A}} [Q_{\boldsymbol{\theta}^{\star}}(s)]_{a}$, with no exploration noise. The same loop scales from this small ticker-picker problem to image-based control without changing the algorithm.

## Advanced: REINFORCE for Continuous Elasticity
The per-regime bandit treats elasticity selection as $K$ stateless single-step problems with discrete arms. That has two structural costs. First, the only state the bandit sees is the regime tag $\rho \in \{\text{bear}, \text{neutral}, \text{bull}\}$, so it cannot use richer signals such as the current drawdown, realized volatility, or the news-flag count, even when those signals are available at decision time. Second, the reward is a single-step pull, so there is no temporal credit assignment: the bandit cannot reward an elasticity choice for the discounted log-growth it produces over the next 30 rebalance days, only for the immediate one-step return. Lifting both restrictions points us at a different RL family: we learn a parametrized policy $\pi_{\boldsymbol{\theta}}(a \mid s)$ that maps a state vector $s$ to a distribution over a continuous action $a$, and we update $\boldsymbol{\theta}$ by gradient ascent on the expected discounted return. This subsection states the framework and instantiates it for elasticity selection; the companion example notebook trains the policy and reports its held-out performance against the bandit and the heuristic.

> __Policy-Gradient Theorem (REINFORCE with Baseline):__
>
> Let $\pi_{\boldsymbol{\theta}}(a \mid s)$ be a parametrized stochastic policy and $J(\boldsymbol{\theta}) = \mathbb{E}_{\tau \sim \pi_{\boldsymbol{\theta}}}\!\left[\sum_{t} \gamma^{t} r_{t}\right]$ the expected discounted return along trajectories $\tau$. The gradient of $J$ admits a Monte Carlo expression that the environment dynamics drop out of:
>
> $$\nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta}) \;=\; \mathbb{E}_{\tau \sim \pi_{\boldsymbol{\theta}}}\!\left[\sum_{t=0}^{T} \nabla_{\boldsymbol{\theta}} \log \pi_{\boldsymbol{\theta}}(a_{t} \mid s_{t}) \cdot G_{t}\right]$$
>
> where $G_{t} = \sum_{k \ge 0} \gamma^{k} r_{t+k}$ is the discounted *future* return from step $t$. Drawing $B$ on-policy trajectories and subtracting any state-only baseline $V_{\boldsymbol{\phi}}(s)$ (which preserves the gradient because $\mathbb{E}_{a \sim \pi_{\boldsymbol{\theta}}}[\nabla_{\boldsymbol{\theta}} \log \pi_{\boldsymbol{\theta}}(a \mid s)] = 0$ pointwise in $s$) gives the **REINFORCE-with-baseline** estimator:
>
> $$\boxed{\;\widehat{\nabla_{\boldsymbol{\theta}} J} \;=\; \frac{1}{B} \sum_{b=1}^{B} \sum_{t=0}^{T} \nabla_{\boldsymbol{\theta}} \log \pi_{\boldsymbol{\theta}}\!\left(a_{t}^{(b)} \mid s_{t}^{(b)}\right) \cdot A_{t}^{(b)}, \qquad A_{t} = G_{t} - V_{\boldsymbol{\phi}}(s_{t})\quad\blacksquare\;}$$
>
> $A_{t}$ is the **advantage**, the realized return minus the baseline's expected return at $s_{t}$. The baseline is unbiased and variance-reducing; the value function is the optimal scalar choice and a learned linear head $V_{\boldsymbol{\phi}}(s) = \boldsymbol{\phi}^{\top} s$ trained by Monte Carlo regression on $G_{t}$ is the simplest version. Original results: [Williams (1992)](https://doi.org/10.1007/BF00992696); modern treatment: [Sutton and Barto (2018), ch. 13](http://incompleteideas.net/book/the-book-2nd.html). The full derivation, including the log-derivative trick and the baseline-bias proof, is in the companion [policy-gradient derivation](deeper/eCornell-AI-Finance-S3-Derivation-PolicyGradient-May-2026.ipynb).

The theorem turns "increase the log-probability of actions that led to high return, decrease the log-probability of actions that led to low return" into an estimator we can compute by sampling. We instantiate the framework for CES elasticity selection below, then pseudocode the full update.

### Applying REINFORCE to CES Elasticity

We instantiate the framework by fixing the state, action, reward, and policy parametrization:

> __Policy-Gradient Setup:__
>
> * **State** $s_t \in \mathbb{R}^{n_s}$: a fixed-order vector of state features. The companion example uses $n_s = 8$ with entries $(\lambda_t^{\text{short}}, \lambda_t^{\text{long}}, \mathbb{1}[\rho_t = \text{bull}], \mathbb{1}[\rho_t = \text{neutral}], \mathbb{1}[\rho_t = \text{bear}], \text{drawdown}_t, \text{realized\_vol}_t, \text{news\_flags}_t)$. The bandit's regime-only state is the special case where the first two and last three entries are zeroed out.
> * **Action** $a_t = \eta_t \in [\eta_{\min}, \eta_{\max}]$: the CES elasticity used by the allocator at step $t$. Continuous, not gridded.
> * **Policy** $\pi_{\boldsymbol{\theta}}(\eta \mid s) = \mathcal{N}(\mu_{\boldsymbol{\theta}}(s), \sigma_{\boldsymbol{\theta}}^2)$: a Gaussian whose mean is linear in the state and whose standard deviation is state-independent. The mean head is $\mu_{\boldsymbol{\theta}}(s) = \mathbf{w}_\mu^\top s + b_\mu$ with parameters $\boldsymbol{\theta} = (\mathbf{w}_\mu, b_\mu, \log \sigma)$. Sampled actions are clamped to $[\eta_{\min}, \eta_{\max}]$ at use time; the gradient is computed against the unsquashed Gaussian draw.
> * **Reward** $r_t$: the realized one-period log-wealth growth $\log(W_{t+1} / W_t)$ produced by stepping the engine forward at $\eta_t$, optionally minus a drawdown penalty $\kappa \cdot \max(0, \text{dd}_t - \text{dd}_{\text{threshold}})$ that lets us shape the policy toward drawdown discipline.
> * **Baseline** $V_{\boldsymbol{\phi}}(s) = \boldsymbol{\phi}^\top s$: a linear value head with parameters $\boldsymbol{\phi} \in \mathbb{R}^{n_s}$, trained by Monte Carlo regression on $G_t$.

For our specific Gaussian policy the gradient of the log-density has closed form: $\nabla_{\mathbf{w}_\mu} \log \pi = (\eta - \mu)/\sigma^2 \cdot s$, $\nabla_{b_\mu} \log \pi = (\eta - \mu)/\sigma^2$, and $\nabla_{\log \sigma} \log \pi = -1 + (\eta - \mu)^2/\sigma^2$. The mean head is pulled toward sampled actions whose advantage is positive and away from those whose advantage is negative; the log-standard-deviation grows when actions further from the mean produced positive advantage (exploration pays) and shrinks when they did not (exploitation tightens).

### Algorithm: REINFORCE-with-Baseline (Online)

__Initialize__: Given a state-feature builder $\mathrm{state}(\cdot)$ that returns $s_t \in \mathbb{R}^{n_s}$ from current engine context, a horizon $T$, an episode batch size $B$, a discount $\gamma \in [0, 1]$, an optimizer step size $\eta_{\mathrm{lr}}$, a baseline step size $\eta_{\boldsymbol{\phi}}$, and an environment that returns realized reward $r_t = \log(W_{t+1}/W_t)$ when the engine runs at elasticity $\eta_t$, initialize the policy parameters $\boldsymbol{\theta} = (\mathbf{w}_\mu, b_\mu, \log \sigma)$ near a neutral midpoint of the action range and the baseline parameters $\boldsymbol{\phi} \gets 0$.

For each Adam update $u = 1, 2, \ldots$ __do__:

1. __Episode rollout__: For $b = 1, \ldots, B$:
    - Sample a starting time $t_0$ uniformly in the valid window.
    - For $t = t_0, t_0 + 1, \ldots, t_0 + T - 1$:
        - Build $s_t \gets \mathrm{state}(\cdot)$ from the current engine context.
        - Sample $\eta_t \sim \mathcal{N}(\mu_{\boldsymbol{\theta}}(s_t), \sigma^2)$ and clamp to $[\eta_{\min}, \eta_{\max}]$.
        - Step the engine forward at $\eta_t$, observe $r_t$, advance shares + cash.
    - Form the per-step returns $G_t^{(b)} \gets \sum_{k \ge 0} \gamma^k r_{t+k}^{(b)}$ by a backward sweep.
2. __Advantage and gradient__: For each $(s_t^{(b)}, \eta_t^{(b)}, G_t^{(b)})$:
    - Read the baseline $V_t \gets \boldsymbol{\phi}^\top s_t^{(b)}$.
    - Form the advantage $A_t \gets G_t^{(b)} - V_t$.
    - Accumulate the policy gradient $\widehat{g} \mathrel{+}= \nabla_{\boldsymbol{\theta}} \log \pi_{\boldsymbol{\theta}}(\eta_t^{(b)} \mid s_t^{(b)}) \cdot A_t$ using the closed-form components above.
    - Update the baseline by gradient descent on its squared error: $\boldsymbol{\phi} \gets \boldsymbol{\phi} - \eta_{\boldsymbol{\phi}} \, (V_t - G_t^{(b)}) \, s_t^{(b)}$.
3. __Adam ascent__: Average $\widehat{g}$ across the $B$ episodes and step the policy parameters $\boldsymbol{\theta} \gets \boldsymbol{\theta} + \eta_{\mathrm{lr}} \cdot \mathrm{Adam}(\widehat{g})$.

__Output__: Return the trained policy parameters $\boldsymbol{\theta}^{\star} = (\mathbf{w}_\mu^{\star}, b_\mu^{\star}, \log \sigma^{\star})$, the trained baseline $\boldsymbol{\phi}^{\star}$, and the per-update learning curves (mean episode return $\bar{G}_0$, gradient norm $\|\widehat{g}\|_2$, mean advantage $|A_t|$).

The deployment-time policy is deterministic: at production time we read the current state $s$ and use the squashed mean $\eta \gets \mathrm{clamp}(\mathbf{w}_\mu^\top s + b_\mu, \eta_{\min}, \eta_{\max})$, with no exploration noise. Production replaces only the elasticity-selection rule that the engine consumes; the engine's compliance gates (drawdown, turnover, concentration) remain unchanged.

## News into the Preference Score

Session 2 declared the news-aware preference weight $\gamma_i = \tanh(\alpha_i / \beta_i^{\lambda_t} + \beta_i^{1-\lambda_t}\,\mathbb{E}[g_{\mathrm{mkt}}] + \nu_i\,\text{news}_i)$, but it left two slots empty: where the per-ticker, per-day news scalar $\text{news}_{i,t}$ comes from at production time, and how the loading $\nu_i$ is estimated from data. This section fills both. The pipeline routes each headline through an LLM to recover a numeric sentiment score, aggregates the per-ticker scores into a daily factor, and recovers $\hat\nu_i$ as one regression coefficient on a held-out training window. The SIM coefficients $(\alpha_i, \beta_i)$ themselves are the next section's responsibility (EWLS); this section is about getting $\hat\nu_i$ and the live $\text{news}_{i,t}$ feed.

> __The Pipeline:__
>
> A scheduled job drives the news flow through three stages:
>
> * __Headline → numeric score.__ Each headline carries the article text, the ticker referenced, the publication day, and a sentiment score $\text{score}_{h} \in [-1, +1]$ assigned by an LLM. The _classifier_ is a zero-shot prompt: we hand Claude the headline text under an instruction asking for one number on $[-1, +1]$ for the named company, with $-1$ strongly bearish, $0$ neutral, $+1$ strongly bullish, and no preamble; the reply is parsed as a `Float64` and clipped to the interval. There is no fine-tuned head, no labeled training set, and no gradient step in the loop: the calibration of the scale and the company-specific reading both live inside the prompt and the model's pretrained weights. Session 4 implements this step as [`score_news_with_claude!`](https://varnerlab.org/eCornell-AI-finance-lectures/dev/session4/#eCornellAIFinance.score_news_with_claude!), and a cached run substitutes calibrated Gaussian noise around the planted signal so the downstream regression sees a realistic LLM error budget without an API call. The bounded output range matters: it lets the regression treat news as a standardized covariate without ad-hoc rescaling.
>
> * __Per-ticker, per-day aggregation.__ Average all sentiment scores for ticker $i$ on day $t$ to build the $T \times K$ news factor matrix:
>     $$\text{news}_{i,t} \;=\; \frac{1}{|H_{i,t}|}\sum_{h \in H_{i,t}} \text{score}_{h}, \qquad \text{news}_{i,t} = 0 \text{ if } H_{i,t} = \emptyset$$
>     where $H_{i,t}$ is the set of headlines about ticker $i$ published on day $t$. Days without coverage receive zero, interpreted as _no news today, no signal_.
>
> * __News-aware SIM regression.__ The Session 1 design $[1,\; g_{m,t}]$ extends to $[1,\; g_{m,t},\; \text{news}_{i,t}]$, and $\hat\nu_i$ is the third coefficient of the resulting OLS estimate.

The closed-form regression is the standard OLS estimator. Stack the per-day rows into the design matrix $\mathbf{X}_i \in \mathbb{R}^{T \times 3}$ with rows $\mathbf{x}_{t} = [1,\; g_{m,t},\; \text{news}_{i,t}]^{\top}$ and the response $\mathbf{y}_i = [g_{i,1}, \ldots, g_{i,T}]^{\top} \in \mathbb{R}^{T}$. The least-squares estimate $\hat{\boldsymbol{\theta}}_i = [\hat\alpha_i^{0},\; \hat\beta_i^{0},\; \hat\nu_i]^{\top}$ is given by:

$$\boxed{\;\hat{\boldsymbol{\theta}}_i \;=\; (\mathbf{X}_i^{\top}\mathbf{X}_i)^{-1}\,\mathbf{X}_i^{\top}\mathbf{y}_i\quad\blacksquare\;}$$

Three quantities come out, only one is taken forward. The third coefficient, $\hat\nu_i$, is the news loading we wanted; it gets frozen and is read by every fire of the engine for the life of the deployment. The first two, $(\hat\alpha_i^{0}, \hat\beta_i^{0})$, are *seed* values: useful as the prior the EWLS recursion in the next section starts from, but not the values the engine reads at production time. EWLS replaces them bar-by-bar.

### Online vs Batch: Two Cadences in One Engine

The asymmetry is intentional. Market exposure (beta) genuinely changes with regime: equity correlations rise in stress, sector concentrations shift after rotations, and a stock's $\hat\beta_i$ a year ago is not its $\hat\beta_i$ today. EWLS is designed for that. A stock's *sensitivity to NLP-extracted sentiment* is more structural: it depends on how often the company shows up in the news cycle, how analysts cover it, and how the disclosure regime treats it. Those drift on the order of months to years, not days. Estimating $\hat\nu_i$ once and freezing it is the simpler, cheaper, and more honest answer; a news-augmented EWLS recursion that updates $\hat\nu_i$ online is a real extension we leave for follow-up.

What refreshes hourly is the news factor itself, $\text{news}_{i,t}$, not the loading $\hat\nu_i$. Each fire reads the latest hourly aggregated score per ticker, multiplies it by the frozen $\hat\nu_i$, and drops the result inside the $\tanh$ of $\gamma_i$. Two cadences, one preference weight.

## EWLS: Online Single Index Model (SIM) Parameter Estimation

The Single Index Model (SIM) regression $g_{i} = \alpha_{i} + \beta_{i} \, g_{\mathrm{mkt}} + \varepsilon_{i}$ can be re-estimated online as new data arrives. Exponentially weighted least squares (EWLS) discounts old observations with a decay factor so the estimates track regime shifts without a sliding window or batch re-calibration.

> __EWLS Recursion (Theorem):__
>
> Fix asset $i$. Let $\mathbf{x}_{s} = [1,\; g_{\mathrm{mkt},s}]^{\top}$, $y_{s} = g_{i,s}$, and $\boldsymbol{\theta}_{i} = [\alpha_{i},\; \beta_{i}]^{\top}$. Choose a half-life $h$ in trading days and set the decay factor $\delta = 2^{-1/h} \in (0, 1)$. The EWLS estimate at time $t$, defined as the minimizer of $L_{t}(\boldsymbol{\theta}) = \sum_{s=1}^{t} \delta^{t-s}\,(y_{s} - \mathbf{x}_{s}^{\top}\boldsymbol{\theta})^{2}$, is recovered from three running moments updated each step:
>
> $$\mathbf{A}_{t} \leftarrow \delta\,\mathbf{A}_{t-1} + \mathbf{x}_{t}\mathbf{x}_{t}^{\top}, \qquad \mathbf{b}_{t} \leftarrow \delta\,\mathbf{b}_{t-1} + \mathbf{x}_{t} y_{t}, \qquad c_{t} \leftarrow \delta\,c_{t-1} + y_{t}^{2}$$
>
> with the parameter estimate and residual scale given by:
>
> $$\hat{\boldsymbol{\theta}}_{i,t} = \begin{bmatrix} \hat{\alpha}_{i,t} \\ \hat{\beta}_{i,t} \end{bmatrix} = \mathbf{A}_{t}^{-1}\,\mathbf{b}_{t}, \qquad \hat{\sigma}_{\varepsilon,i,t} = \sqrt{\frac{c_{t} - \hat{\boldsymbol{\theta}}_{i,t}^{\top}\,\mathbf{b}_{t}}{[\mathbf{A}_{t}]_{1,1}}}\quad\blacksquare$$
>
> The matrix $\mathbf{A}_{t} \in \mathbb{R}^{2 \times 2}$, vector $\mathbf{b}_{t} \in \mathbb{R}^{2}$, and scalar $c_{t} \in \mathbb{R}$ are sufficient statistics: the entire weighted history collapses into these three running quantities, updated by a single decayed-plus-rank-one step per day. The full proof is in the companion [EWLS recursion derivation notebook](deeper/eCornell-AI-Finance-S3-Derivation-EWLS-Recursion-May-2026.ipynb).

The recursion runs on every asset in parallel and consumes constant memory and constant compute per trading day. We translate it directly into pseudocode below.

### Algorithm: EWLS SIM Update (Online)

__Initialize__: Given a calibrated prior $(\boldsymbol{\theta}_{i,0}, \sigma_{\varepsilon,i,0})$ for each asset $i$, a prior weight $W_{0} \geq 0$, a half-life $h$ in trading days with decay $\delta = 2^{-1/h}$, and a stream of observations $\{(\mathbf{x}_{t}, y_{t})\}_{t=1}^{T}$ with $\mathbf{x}_{t} = [1,\; g_{\mathrm{mkt},t}]^{\top}$ and $y_{t} = g_{i,t}$, seed the running moments so that $W_{0}$ pseudo-observations carry the prior:
$$\mathbf{A}_{0} \gets W_{0} \cdot \mathbb{E}[\mathbf{x}\mathbf{x}^{\top}], \qquad \mathbf{b}_{0} \gets \mathbf{A}_{0}\,\boldsymbol{\theta}_{i,0}, \qquad c_{0} \gets \boldsymbol{\theta}_{i,0}^{\top}\mathbf{b}_{0} + W_{0}\,\sigma_{\varepsilon,i,0}^{2}$$

For $t = 1, \ldots, T$ __do__:

1. Age the existing moments by the decay factor: $\mathbf{A}_{t} \gets \delta\,\mathbf{A}_{t-1}$, $\mathbf{b}_{t} \gets \delta\,\mathbf{b}_{t-1}$, $c_{t} \gets \delta\,c_{t-1}$.
2. Fold in today's observation $(\mathbf{x}_{t}, y_{t})$:
    $$\mathbf{A}_{t} \gets \mathbf{A}_{t} + \mathbf{x}_{t}\mathbf{x}_{t}^{\top}, \qquad \mathbf{b}_{t} \gets \mathbf{b}_{t} + \mathbf{x}_{t} y_{t}, \qquad c_{t} \gets c_{t} + y_{t}^{2}$$
3. Solve the $2 \times 2$ weighted normal equations for the parameter estimate: $\hat{\boldsymbol{\theta}}_{i,t} \gets \mathbf{A}_{t}^{-1}\,\mathbf{b}_{t}$.
4. Compute the residual standard deviation estimate: $\hat{\sigma}_{\varepsilon,i,t} \gets \sqrt{(c_{t} - \hat{\boldsymbol{\theta}}_{i,t}^{\top}\,\mathbf{b}_{t}) / [\mathbf{A}_{t}]_{1,1}}$.
5. Hand $(\hat{\alpha}_{i,t}, \hat{\beta}_{i,t}, \hat{\sigma}_{\varepsilon,i,t})$ to the engine for use at the next rebalance.

__Output__: Return the running SIM estimates $\{(\hat{\alpha}_{i,t}, \hat{\beta}_{i,t}, \hat{\sigma}_{\varepsilon,i,t})\}_{t=1}^{T}$ and the final sufficient statistics $(\mathbf{A}_{T}, \mathbf{b}_{T}, c_{T})$.

The recursion runs once per asset, in parallel across the universe; only the prior-weighting matrix $\mathbf{A}_{0}$ and the decay factor $\delta$ are shared across assets.

### Prior Seeding and Half-Life

Two knobs control how aggressively EWLS adapts: the prior weight $W_{0}$ that anchors the recursion before any new data arrives, and the half-life $h$ that sets how fast old data fades.

* __Prior weight $W_{0}$:__ Before any observation is processed, the sufficient statistics are seeded as if $W_{0}$ pseudo-observations consistent with the calibrated parameters $\boldsymbol{\theta}_{i,0} = [\alpha_{i,0},\; \beta_{i,0}]^{\top}$ and residual scale $\sigma_{\varepsilon,i,0}$ had been seen, giving $\mathbf{A}_{0} \approx W_{0} \cdot \mathbb{E}[\mathbf{x}\mathbf{x}^{\top}]$, $\mathbf{b}_{0} \approx W_{0} \cdot \mathbb{E}[\mathbf{x}\mathbf{x}^{\top}]\,\boldsymbol{\theta}_{i,0}$, and $c_{0}$ accordingly. Large $W_{0}$ keeps the calibrated values dominant longer; small $W_{0}$ lets the first few real observations move the estimates appreciably.
* __Half-life $h$ (speed):__ A short half-life such as $h = 21$ days (approximately one trading month) puts most of the weight on the last few weeks of data, so the estimates track regime shifts quickly but inherit single-day noise. The $h = 21$ setting is a good choice when responsiveness matters more than smoothness.
* __Half-life $h$ (stability):__ A long half-life such as $h = 126$ days (six months) averages over a much wider window and produces smooth estimates that lag genuine regime transitions. The default $h = 63$ days (one trading quarter) sits between these extremes and is the value the engine ships with for daily rebalancing.

Together $W_{0}$ and $h$ move the estimator along the speed-stability frontier; everything else in the recursion is mechanical.

## Validation and Compliance

Before the engine goes to production in Session 4, it must pass a formal validation. The validation report applies five pass/fail gates to the Monte Carlo backtest results:

| Criterion | Threshold | Rationale |
|-----------|-----------|-----------|
| **Median [Sharpe](../session-1/eCornell-AI-Finance-S1-Lecture-StressTestingMinVariancePortfolios-May-2026.ipynb#the-capital-allocation-line-and-tangent-portfolio)** $\geq$ **0.3** | Risk-adjusted growth must exceed a low bar | Below 0.3, the strategy is not compensating for the risk it takes |
| **Median [Max Drawdown](../session-1/eCornell-AI-Finance-S1-Lecture-StressTestingMinVariancePortfolios-May-2026.ipynb#portfolio-npv-and-tail-risk-metrics)** $\leq$ **25%** | Worst-case capital loss must be bounded | Larger drawdowns trigger investor redemptions and regulatory scrutiny |
| **Failure Rate** $\leq$ **10%** | At most 10% of paths end below initial capital | Systematic losses indicate a flawed strategy |
| **Wealth versus Frozen Baseline** $\geq$ **1.0** | Median terminal wealth must beat the frozen-parameter engine | An online learner that does not beat its frozen ancestor is not earning its complexity |
| **Median NPV** $\geq$ **0** | Median path must beat the risk-free baseline in present-value terms | A strategy whose median NPV is negative is dominated by a T-bill |

A strategy that passes all five gates across model-generated paths has demonstrated robustness to regime shifts, fat tails, and volatility clustering. It is not guaranteed to work in production, but it has cleared the minimum bar for deployment.

The validation report also produces a set of **compliance parameters** that become the pre-approved operating limits for Session 4's production system:

* **Concentration cap:** maximum portfolio weight in any single asset (e.g., 40%)
* **Drawdown gate:** maximum drawdown before the engine de-risks to cash (e.g., 15%)
* **Turnover limit:** maximum daily trade volume as a fraction of portfolio value (e.g., 50%)
* **Position size limit:** maximum dollar exposure per individual trade (e.g., $5,000)

In Session 4, trades within these limits auto-execute; trades that exceed them are queued for human review. The compliance configuration is the bridge between _validated in simulation_ and _approved for production._

## Summary

In this session, we filled the news pipeline behind the Session 2 news-aware $\gamma_i$, replaced frozen SIM parameters with online estimation, replaced the hand-picked universe with a learned subset over $2^{K}-1$ candidate baskets, and validated the result against formal deployment gates.

> __Key Takeaways:__
>
> * __Reinforcement learning frames portfolio allocation as sequential decision-making:__ At each rebalance the engine is an agent in a state $s$ choosing an action $a$, observing a reward $r$, and updating its policy. Different RL families (multiplicative weights, bandits, Q-learning) suit different action-space and feedback structures.
> * __Epsilon-greedy bandits balance exploration and exploitation:__ At each round the bandit explores a random arm with decaying probability $\varepsilon_t = t^{-1/3}(K \ln t)^{1/3}$ and otherwise pulls the arm with the highest estimated mean. The schedule achieves provably sublinear regret: the learner converges on the best arm while never stopping exploration entirely.
> * __Bandits learn the tradeable universe:__ Applied to the $2^{K}-1$ subsets of a $K$-ticker candidate pool with Cobb-Douglas utility as the reward, the bandit replaces the hand-picked Session 1 universe with a data-driven subset that adapts as preferences and prices change.
> * __Validation gates enforce deployment readiness:__ Five pass/fail criteria (Sharpe, drawdown, failure rate, wealth versus the frozen baseline, and median NPV) produce a binary go/no-go decision for production deployment. The compliance configuration exports the operating limits that Session 4's two-tier execution system enforces in real time.

**Next Session:** In [Session 4](../session-4/eCornell-AI-Finance-S4-Lecture-ProductionOps-May-2026.ipynb), the validated engine moves to production with three operational layers:

* __Cron execution:__ A scheduled agent runs the engine during market hours, reading the same compliance limits defined in this session's validation report and submitting orders against the live broker.
* __Two-tier policy:__ Trades that fall inside the pre-approved compliance envelope (concentration, drawdown gate, turnover, position size) auto-execute; trades that exceed any limit queue for human review before they can be released.
* __Post-market review:__ The evening session walks the day's audit trail, P&L attribution, and escalation log so every order, fill, and queued proposal is accounted for before the next open.

Friday's compliance config from Example 3 is the artifact that makes all three layers possible.
